## Importar y crear objetos
`import flet as ft` carga el módulo y crea alias. `from views.population_editor import PopulationEditor` importa una clase local. Python incluye src en la búsqueda cuando ejecutas src/main.py.

`ft.Text('Hola', size=24)` crea una instancia; 'Hola' es argumento posicional y size argumento nombrado. Una clase derivada, como PopulationEditor(ft.Container), hereda comportamiento del contenedor. `super().__init__` inicializa la base. `self` identifica la instancia. `@property` expone un dato mediante acceso como editor.individuals.

`tuple[float,...]` es un vector de longitud variable; list almacena elementos editables; tuple es inmutable. Las anotaciones no validan automáticamente entradas. dataclass genera constructor a partir de campos; frozen=True impide reasignarlos. Protocol define métodos esperados sin obligar a heredar de él.

## Página y callbacks
ft.run(main) inicia la app y entrega Page. page.add monta controles; cambiar una propiedad modifica el objeto; page.update comunica cambios. El guard __name__ evita iniciar la app al importar módulos.

on_click recibe una función sin ejecutarla; saludar() ejecutaría inmediatamente. e es el evento, aunque no siempre se necesita leerlo. Las lambdas de filas guardan i=index para que cada botón recuerde su propio individuo.

## Ejemplo opcional
La siguiente celda abre una ventana sólo si descomentas ft.run. Ciérrala para terminar.


```python
import flet as ft

def ejemplo(page: ft.Page):
    texto = ft.Text('Hola, Flet', size=24)
    def saludar(e):
        texto.value = 'Has pulsado el botón'
        page.update()
    page.add(ft.Column([texto, ft.Button('Saludar', on_click=saludar)]))

# ft.run(ejemplo)
```

Text.value es texto; size tamaño; Column.controls lista vertical; Button.content etiqueta; on_click acción. try/except captura errores y raise ValueError los comunica. async def crea coroutine; await cede ejecución. Consulta [Flet](https://flet.dev/docs/).
