## Tipos y parámetros
Individual = tuple[float,...]. En 1D un individuo es `(x,)`, con coma para crear tupla; en 2D es `(x,y)`.

Objective(name,expression,labels,bounds,evaluate): nombre, fórmula, etiqueta por coordenada, pares (mínimo,máximo) y callable que recibe el vector completo. dimensions devuelve len(labels).

AlgorithmConfig(population,objective,mutation_rate,selection,generations): población inmutable aplicada, objetivo, probabilidad [0,1], clave torneo/ruleta/elitista y número de iteraciones. La población determina su tamaño; el campo Cantidad para generar sólo configura la muestra automática.

Generation(index,population,scores): índice consecutivo desde 0, vectores y puntuación por vector. EvolutionEngine.run(config) devuelve una lista de estas generaciones.

## Funciones provisionales
| Clave | Dominio por variable | Evaluación |
|---|---|---|
| sphere | −5 a 5 | suma de cuadrados |
| rastrigin | −5.12 a 5.12 | 10d + suma(v²−10cos(2πv)) |
| ackley | −5 a 5 | fórmula de Ackley con medias de cuadrados y cosenos |
| quartic | −2 a 4 | suma(v⁴−3v³+2) |
| wave | −6 a 6 | suma(sin(3v)+0.1v²) |

get_objective(key,dimensions) crea versión de 1 o 2 variables con etiquetas x,y. Todos los ejemplos se minimizan. math provee exp/sin/cos/pi/e, y typing Callable expresa la firma evaluadora.

## Validación y demostración
validate_population exige 2–200 individuos, cantidad correcta de coordenadas, números finitos y dominio válido. validate_generations exige generación 0 y todas las iteraciones, índices consecutivos, tamaño constante y scores finitos.

initial_population(size,objective) usa random.Random(42) y uniform por coordenada; es reproducible. DemoEngine.run muestrea 61 puntos por eje, combina con itertools.product y busca el mínimo aproximado. Interpola cada individuo hacia esa posición. No garantiza mejora monótona y no utiliza selección ni mutación.


```python
from pathlib import Path
import sys
candidates = [Path.cwd(), *Path.cwd().parents]
project = next((base/'AM' for base in candidates if (base/'AM'/'src').is_dir()), None)
if project is None:
    project = next(base for base in candidates if (base/'src').is_dir())
sys.path.insert(0, str(project/'src'))
from models.objectives import get_objective
from models.demo_engine import initial_population, DemoEngine
from models.contracts import AlgorithmConfig, validate_population, validate_generations
objective = get_objective('sphere', 2)
population = validate_population(initial_population(5, objective), objective)
config = AlgorithmConfig(population, objective, 0.05, 'torneo', 3)
result = DemoEngine().run(config)
validate_generations(result, config)
assert result[0].population == population
print(result[-1])
```
