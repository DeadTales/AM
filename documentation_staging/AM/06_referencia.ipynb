### `dimensions`
```python
@property
```
Número de coordenadas de cada individuo.

Parámetros: self. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

### `run`
```python
def run(self, config: AlgorithmConfig) -> list[Generation]:
```
Devuelve generaciones 0..config.generations; 0 es la población inicial.

Parámetros: self, config. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Contratos independientes de Flet para conectar cualquier motor genético."""
from dataclasses import dataclass
import math
from typing import Callable, Protocol

Individual = tuple[float, ...]


@dataclass(frozen=True)
class Objective:
    """Nombre, fórmula, etiquetas, dominio por coordenada y evaluación del vector."""
    name: str
    expression: str
    labels: tuple[str, ...]
    bounds: tuple[tuple[float, float], ...]
    evaluate: Callable[[Individual], float]

    @property
    def dimensions(self) -> int:
        """Número de coordenadas de cada individuo."""
        return len(self.labels)


@dataclass(frozen=True)
class AlgorithmConfig:
    """Entrada al motor: población, objetivo, probabilidad [0,1], selección e iteraciones."""
    population: tuple[Individual, ...]
    objective: Objective
    mutation_rate: float
    selection: str
    generations: int


@dataclass(frozen=True)
class Generation:
    """Salida: índice desde cero, población y una puntuación por individuo."""
    index: int
    population: tuple[Individual, ...]
    scores: tuple[float, ...]


class EvolutionEngine(Protocol):
    """Implementar run(config)"""

    def run(self, config: AlgorithmConfig) -> list[Generation]:
        """Devuelve generaciones 0..config.generations; 0 es la población inicial."""
        ...


def validate_population(population, objective: Objective) -> tuple[Individual, ...]:
    """Normaliza vectores y rechaza cantidad, dimensión, valores o dominios inválidos."""
    if not 2 <= len(population) <= 200:
        raise ValueError("La población debe contener entre 2 y 200 individuos.")
    result = []
    for index, individual in enumerate(population):
        if len(individual) != objective.dimensions:
            raise ValueError(f"El individuo {index+1} requiere {objective.dimensions} coordenadas.")
        try:
            values = tuple(float(value) for value in individual)
        except (TypeError, ValueError):
            raise ValueError(f"El individuo {index+1} contiene un valor no numérico.") from None
        for label, value, (lower, upper) in zip(objective.labels, values, objective.bounds):
            if not math.isfinite(value) or not lower <= value <= upper:
                raise ValueError(f"Individuo {index+1}: {label} debe estar entre {lower:g} y {upper:g}.")
        result.append(values)
    return tuple(result)


def validate_generations(generations: list[Generation], config: AlgorithmConfig) -> None:
    """Verifica salida del motor antes de mostrarla, incluida la generación inicial."""
    if len(generations) != config.generations+1:
        raise ValueError("El motor debe devolver la generación 0 y todas las iteraciones solicitadas.")
    for index, generation in enumerate(generations):
        if generation.index != index:
            raise ValueError("Los índices de generación deben ser consecutivos desde cero.")
        population = validate_population(generation.population, config.objective)
        if len(population) != len(config.population) or len(generation.scores) != len(population):
            raise ValueError("El motor debe conservar el tamaño de población y devolver una puntuación por individuo.")
        if any(not math.isfinite(score) for score in generation.scores):
            raise ValueError("El motor devolvió puntuaciones no finitas.")
    if generations[0].population != config.population:
        raise ValueError("La generación 0 debe contener la población inicial aplicada.")
````

### `run`
```python
def run(self, config: AlgorithmConfig) -> list[Generation]:
```
Interpola hacia un mínimo muestreado; selección y mutación no actúan.

Parámetros: self, config. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Motor ilustrativo reemplazable; no implementa selección, cruce ni mutación."""
import itertools
import random
from models.contracts import AlgorithmConfig, Generation, Individual, Objective


def initial_population(size: int, objective: Objective) -> list[Individual]:
    """Genera size vectores reproducibles dentro de los límites de cada coordenada."""
    if not 2 <= size <= 200:
        raise ValueError("La cantidad para generar debe estar entre 2 y 200.")
    rng = random.Random(42)
    return [tuple(rng.uniform(lower, upper) for lower, upper in objective.bounds) for _ in range(size)]


class DemoEngine:
    """Implementación ilustrativa de EvolutionEngine que puedes reemplazar por tu motor."""

    def run(self, config: AlgorithmConfig) -> list[Generation]:
        """Interpola hacia un mínimo muestreado; selección y mutación no actúan."""
        objective = config.objective
        axes = [[lower+(upper-lower)*i/60 for i in range(61)] for lower, upper in objective.bounds]
        target = min(itertools.product(*axes), key=objective.evaluate)
        result = []
        for index in range(config.generations+1):
            progress = index/config.generations
            population = config.population if index == 0 else tuple(tuple(v+(t-v)*progress for v, t in zip(individual, target)) for individual in config.population)
            result.append(Generation(index, population, tuple(objective.evaluate(individual) for individual in population)))
        return result
````

### `evaluate`
```python
def evaluate(values: Individual) -> float:
```
Evalúa el vector completo; todos los ejemplos se minimizan.

Parámetros: values. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Registro de cinco funciones provisionales de una o dos variables."""
import math
from models.contracts import Individual, Objective

FUNCTIONS = {"sphere": "Esfera", "rastrigin": "Rastrigin", "ackley": "Ackley", "quartic": "Cuártica", "wave": "Ondulada"}


def get_objective(key: str, dimensions: int) -> Objective:
    """Crea objetivo por clave y número de variables, etiquetadas x e y."""
    if dimensions not in (1, 2):
        raise ValueError("Las funciones de ejemplo admiten una o dos variables.")
    labels = ("x", "y")[:dimensions]
    domains = {"sphere": (-5, 5), "rastrigin": (-5.12, 5.12), "ackley": (-5, 5), "quartic": (-2, 4), "wave": (-6, 6)}
    expressions = {"sphere": "Σ vᵢ²", "rastrigin": "10d + Σ [vᵢ² − 10 cos(2πvᵢ)]", "ackley": "−20 exp(−0.2 √media(vᵢ²)) − exp(media(cos(2πvᵢ))) + 20 + e", "quartic": "Σ [vᵢ⁴ − 3vᵢ³ + 2]", "wave": "Σ [sin(3vᵢ) + 0.1vᵢ²]"}

    def evaluate(values: Individual) -> float:
        """Evalúa el vector completo; todos los ejemplos se minimizan."""
        if key == "sphere":
            return sum(v*v for v in values)
        if key == "rastrigin":
            return 10*dimensions+sum(v*v-10*math.cos(2*math.pi*v) for v in values)
        if key == "ackley":
            return -20*math.exp(-0.2*math.sqrt(sum(v*v for v in values)/dimensions))-math.exp(sum(math.cos(2*math.pi*v) for v in values)/dimensions)+20+math.e
        if key == "quartic":
            return sum(v**4-3*v**3+2 for v in values)
        return sum(math.sin(3*v)+0.1*v*v for v in values)

    return Objective(FUNCTIONS[key], "f("+", ".join(labels)+") = "+expressions[key], labels, (domains[key],)*dimensions, evaluate)
````

### `__init__`

### `set_image`
```python
def set_image(self, png: bytes):
```
Sustituye la imagen sin recrear el visor ni perder el estado del zoom.

Parámetros: self, png. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Gráficas renderizadas a PNG: evitan comparar cientos de primitivas Canvas por actualización."""
from io import BytesIO
from base64 import b64decode
import flet as ft
import numpy as np
from matplotlib.backends.backend_agg import FigureCanvasAgg
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.figure import Figure
from models.contracts import Objective
from views.theme import ACCENT, BORDER, MUTED


class Chart(ft.Container):
    """Visor persistente con dimensiones explícitas; cambiar src no reconstruye sus controles."""

    def __init__(self, height: int = 320):
        """height fija el alto del visor; el ancho se ajusta al espacio de la tarjeta."""
        super().__init__(height=height, width=float("inf"))
        placeholder = b64decode("iVBORw0KGgoAAAANSUhEUgAAAAEAAAABCAQAAAC1HAwCAAAAC0lEQVR42mNk+A8AAQUBAScY42YAAAAASUVORK5CYII=")
        self.raster = ft.Image(src=placeholder, width=800, height=height, fit=ft.BoxFit.CONTAIN, gapless_playback=True, semantics_label="Gráfica de la evolución")
        self.viewer = ft.InteractiveViewer(content=self.raster, height=height, width=float("inf"), constrained=True, min_scale=1, max_scale=4, trackpad_scroll_causes_scale=True)
        self.content = self.viewer

    def set_image(self, png: bytes):
        """Sustituye la imagen sin recrear el visor ni perder el estado del zoom."""
        self.raster.src = png
        self.raster.visible = True


def _figure(height: int = 320):
    """Crea figura independiente de pyplot con estilo claro y margen para etiquetas."""
    figure = Figure(figsize=(8, height/100), dpi=100, layout="constrained", facecolor="white")
    axes = figure.add_subplot(111)
    axes.tick_params(colors=MUTED, labelsize=9)
    for name, spine in axes.spines.items():
        spine.set_color(BORDER)
        spine.set_visible(name not in ("top", "right"))
    return figure, axes


def _png(figure: Figure) -> bytes:
    """Renderiza con Agg sin ventanas auxiliares y devuelve bytes PNG para ft.Image."""
    stream = BytesIO()
    FigureCanvasAgg(figure).print_png(stream)
    result = stream.getvalue()
    figure.clear()
    return result


def plot(curve: list[tuple[float, float]], dots: list[tuple[float, float]], color: str = ACCENT, xlabel: str = "x") -> bytes:
    """Dibuja curva y puntos en un PNG; xlabel nombra el eje horizontal."""
    figure, axes = _figure()
    axes.plot(*zip(*curve), color=color, linewidth=1.8)
    if dots:
        axes.scatter(*zip(*dots), c="#A6633D", s=26, edgecolors="white", linewidths=0.7, zorder=3)
    axes.grid(color=BORDER, linewidth=0.6, alpha=0.7)
    axes.set_xlabel(xlabel, fontsize=10, color=MUTED)
    axes.set_ylabel("Evaluación", fontsize=10, color=MUTED)
    return _png(figure)


def heatmap(objective: Objective, population, best: int) -> bytes:
    """Muestra f(x,y) con color e individuos superpuestos, destacando best."""
    (xmin, xmax), (ymin, ymax) = objective.bounds
    resolution = 40
    xs = [xmin+(xmax-xmin)*(i+0.5)/resolution for i in range(resolution)]
    ys = [ymin+(ymax-ymin)*(j+0.5)/resolution for j in range(resolution)]
    values = [[objective.evaluate((x, y)) for x in xs] for y in ys]
    figure, axes = _figure(340)
    palette = LinearSegmentedColormap.from_list("evolution", ["#EDF2F7", "#53789B"])
    surface = axes.imshow(values, extent=(xmin, xmax, ymin, ymax), origin="lower", aspect="auto", cmap=palette)
    axes.scatter(*zip(*population), c="#243247", s=22, edgecolors="white", linewidths=0.7)
    axes.scatter(*population[best], c="#9C5837", s=65, edgecolors="white", linewidths=1.2, zorder=4)
    axes.set_xlabel(objective.labels[0], fontsize=10, color=MUTED)
    axes.set_ylabel(objective.labels[1], fontsize=10, color=MUTED)
    colorbar = figure.colorbar(surface, ax=axes, pad=0.025)
    colorbar.set_label("f(x, y)", fontsize=9, color=MUTED)
    colorbar.ax.tick_params(labelsize=8, colors=MUTED)
    return _png(figure)


def surface3d(objective: Objective, population, best: int, elevation: float = 30, azimuth: float = -55) -> bytes:
    """Dibuja z=f(x,y), individuos sobre la superficie y cámara en grados."""
    (xmin, xmax), (ymin, ymax) = objective.bounds
    x, y = np.meshgrid(np.linspace(xmin, xmax, 55), np.linspace(ymin, ymax, 55))
    z = np.array([[objective.evaluate((float(a), float(b))) for a, b in zip(row_x, row_y)] for row_x, row_y in zip(x, y)])
    figure = Figure(figsize=(8, 4.2), dpi=100, layout="constrained", facecolor="white")
    axes = figure.add_subplot(111, projection="3d", computed_zorder=False)
    palette = LinearSegmentedColormap.from_list("evolution3d", ["#EDF2F7", "#53789B"])
    axes.plot_surface(x, y, z, cmap=palette, alpha=0.8, linewidth=0, antialiased=True, rcount=55, ccount=55, zorder=1)
    px, py = zip(*population)
    scores = [objective.evaluate(individual) for individual in population]
    axes.scatter(px, py, scores, c="#243247", s=22, edgecolors="white", linewidths=0.5, depthshade=False, zorder=3)
    axes.scatter(*population[best], scores[best], c="#9C5837", s=75, edgecolors="white", linewidths=1, depthshade=False, zorder=4)
    axes.view_init(elev=elevation, azim=azimuth)
    axes.set_xlabel(objective.labels[0], color=MUTED, fontsize=9)
    axes.set_ylabel(objective.labels[1], color=MUTED, fontsize=9)
    axes.set_zlabel("f(x, y)", color=MUTED, fontsize=9)
    axes.tick_params(colors=MUTED, labelsize=8)
    axes.set_box_aspect((1, 1, 0.65))
    return _png(figure)
````

````python
"""Vistas de configuración y resultados, independientes del motor de evolución."""
import asyncio
import flet as ft
from models.contracts import AlgorithmConfig, EvolutionEngine, validate_generations, validate_population
from models.demo_engine import DemoEngine, initial_population
from models.objectives import FUNCTIONS, get_objective
from views.charts import Chart, heatmap, plot, surface3d
from views.population_editor import PopulationEditor
from views.menu import Menu, MenuItem
from views.router import ViewRouter
from views.theme import ACCENT, BORDER, ERROR, MUTED, SOFT, TEXT, card, field


class Dashboard:
    """Coordina configuración, ejecución del motor y navegación por resultados."""

    def __init__(self, page: ft.Page, engine: EvolutionEngine | None = None):
        """page aloja controles; engine implementa run(config), por defecto DemoEngine."""
        self.page = page
        self.engine = engine if engine is not None else DemoEngine()
        self.index, self.total = 0, 40
        self.playing, self.busy, self.revision = False, False, 0
        self.snapshots = []
        self.history = []
        self.config = None
        self.is_demo = isinstance(self.engine, DemoEngine)
        self.function = ft.Dropdown(label="Función objetivo", value="sphere", options=[ft.DropdownOption(key=k, text=name) for k, name in FUNCTIONS.items()], on_select=self.change_objective)
        self.dimensions = ft.Dropdown(label="Variables por individuo", value="2", options=[ft.DropdownOption(key="1", text="Una variable · x"), ft.DropdownOption(key="2", text="Dos variables · x, y")], on_select=self.change_objective)
        self.size = field("Cantidad para generar", "20")
        self.generations = field("Generaciones", "40", on_change=self.invalidate)
        self.selection = ft.Dropdown(label="Selección", value="torneo", options=[ft.DropdownOption(key=k, text=k.capitalize()) for k in ("torneo", "ruleta", "elitista")], on_select=self.invalidate)
        self.mutation_label = ft.Text("Mutación · 5 %", color=TEXT, size=14)
        self.mutation = ft.Slider(min=0, max=100, divisions=100, value=5, label="{value}%", active_color=ACCENT, on_change=self.change_mutation)
        self.message = ft.Text("Ajusta los parámetros y la población; después pulsa Ver resultados.", size=13, color=MUTED)
        self.formula = ft.Text(color=MUTED, size=12)
        self.editor = PopulationEditor(dimensions=2, labels=("x", "y"), bounds=get_objective("sphere", 2).bounds, on_change=self.population_changed, on_refresh=self.page.update)
        self.metrics = ft.ResponsiveRow(spacing=14, run_spacing=12)
        self.function_chart = Chart(height=420)
        self.chart_mode = ft.Dropdown(label="Visualización", value="surface", options=[ft.DropdownOption(key="surface", text="Superficie 3D"), ft.DropdownOption(key="heatmap", text="Mapa de calor")], on_select=self.change_chart)
        self.azimuth = ft.Slider(min=-180, max=180, divisions=72, value=-55, label="Giro: {value}°", active_color=ACCENT, on_change_end=self.change_chart)
        self.elevation = ft.Slider(min=5, max=85, divisions=16, value=30, label="Inclinación: {value}°", active_color=ACCENT, on_change_end=self.change_chart)
        self.camera_controls = ft.ResponsiveRow([ft.Column([ft.Text("Giro horizontal", size=12, color=MUTED), self.azimuth], col={"xs":12,"sm":6}, spacing=0), ft.Column([ft.Text("Inclinación", size=12, color=MUTED), self.elevation], col={"xs":12,"sm":6}, spacing=0)], visible=False)
        self.chart_controls = ft.Column([self.chart_mode, self.camera_controls], visible=False)
        self.history_chart = Chart(height=300)
        self.table = ft.ListView(height=330, spacing=8)
        self.timeline = ft.Slider(min=0, max=40, divisions=40, value=0, label="Generación {value}", active_color=ACCENT, on_change=self.seek)
        self.play_button = ft.Button("Reproducir", icon=ft.Icons.PLAY_ARROW, on_click=self.toggle_play)
        self.progress = ft.ProgressBar(value=0, color=ACCENT, bgcolor=SOFT)
        self.apply_button = ft.Button("Ver resultados", icon=ft.Icons.ARROW_FORWARD, on_click=self.apply)
        self.edit_button = ft.Button("Editar configuración", icon=ft.Icons.TUNE, on_click=lambda e: self.show_view("config"), visible=False)
        self.action_hint = ft.Text("Se usarán los individuos guardados en la lista.", size=12, color=MUTED, expand=True)
        self.actions = ft.Container(content=ft.Row([self.action_hint, self.apply_button, self.edit_button], spacing=12), padding=12, bgcolor=SOFT, border_radius=12)
        self.result_summary = ft.Text(size=13, color=MUTED)
        self.configuration_view = ft.Column()
        self.results_view = ft.Column(visible=False)
        self.router = ViewRouter({"config": self.configuration_view, "results": self.results_view}, initial="config", guards={"results": lambda: bool(self.snapshots)}, on_change=self._on_route_change)
        self.menu = Menu([MenuItem("config", "Configuración", ft.Icons.TUNE), MenuItem("results", "Resultados", ft.Icons.INSIGHTS)], on_navigate=self.show_view, on_back=self.router.back)
        self.menu.set_enabled("results", False)
        self.menu.select("config")
        self._prepare_population()

    def _prepare_population(self):
        """Carga población inicial de ejemplo y fórmula sin ejecutar el motor."""
        objective = get_objective(self.function.value, int(self.dimensions.value))
        self.editor.set_schema(objective.dimensions, objective.labels, objective.bounds, notify=False)
        self.editor.set_individuals(initial_population(int(self.size.value), objective), notify=False)
        self.formula.value = objective.expression+"\n"+" · ".join(f"{label} ∈ [{low:g}, {high:g}]" for label, (low, high) in zip(objective.labels, objective.bounds))

    def build(self) -> ft.Column:
        """Construye dos vistas para separar preparación y análisis visual."""
        self.configuration_view.controls = [ft.ResponsiveRow([
            card("Parámetros", "Define cómo se evaluará tu población.", [self.function, self.dimensions, self.formula, ft.Divider(color=BORDER), self.generations, self.selection, ft.Column([self.mutation_label, self.mutation], spacing=0)], {"xs":12, "lg":4}),
            card("Población inicial", "Cada fila es un individuo. Puedes agregarlo, editarlo o eliminarlo.", [ft.Row([self.size, ft.TextButton("Generar población", icon=ft.Icons.REFRESH, on_click=self.configure)], wrap=True), self.editor], {"xs":12, "lg":8}),
        ], spacing=20, run_spacing=20)]
        self.results_view.controls = [self.result_summary, self.metrics, ft.ResponsiveRow([
            card("Función y población", "Superficie 3D o mapa de calor para dos variables · Zoom y arrastre", [self.chart_controls, self.function_chart], {"xs":12, "lg":8}),
            card("Individuos", "Valores de la generación actual; el mejor aparece destacado.", [self.table], {"xs":12, "lg":4}),
        ], spacing=20, run_spacing=20), card("Recorrido de generaciones", "Eje horizontal: generación · Eje vertical: mejor evaluación observada", [ft.Row([ft.IconButton(icon=ft.Icons.SKIP_PREVIOUS, tooltip="Volver a generación 0", on_click=self.reset), ft.IconButton(icon=ft.Icons.CHEVRON_LEFT, tooltip="Generación anterior", on_click=lambda e: self.step(-1)), self.play_button, ft.IconButton(icon=ft.Icons.CHEVRON_RIGHT, tooltip="Siguiente generación", on_click=lambda e: self.step(1))], wrap=True), self.timeline, self.progress, self.history_chart])]
        self.body = ft.Column([self.message, self.router.outlet], spacing=16, scroll=ft.ScrollMode.AUTO, expand=True, horizontal_alignment=ft.CrossAxisAlignment.STRETCH)
        return ft.Column([ft.Row([ft.Icon(ft.Icons.SCIENCE_OUTLINED, color=ACCENT, size=26), ft.Text("Evolución Lab", size=24, weight=ft.FontWeight.W_600, color=TEXT), ft.Text("Demostración" if self.is_demo else "Motor conectado", color=ACCENT, size=12)], spacing=12), self.menu, self.body, self.actions], spacing=12, expand=True, horizontal_alignment=ft.CrossAxisAlignment.STRETCH)

    def show_view(self, name: str):
        """Solicita navegación al router conservando edición y resultados existentes."""
        return self.router.go(name)

    def _on_route_change(self, name: str):
        """Pausa reproducción y sincroniza el menú cuando el router cambia de vista."""
        self._pause()
        self.menu.select(name, self.router.can_go_back)
        self.apply_button.visible = name == "config"
        self.edit_button.visible = name == "results"
        self.action_hint.value = "Se usarán los individuos guardados en la lista." if name == "config" else "Puedes volver a editar sin perder los resultados."
        self.page.update()
        if hasattr(self.page, "run_task"):
            self.page.run_task(self._scroll_to_start)

    async def _scroll_to_start(self):
        """Vuelve al inicio del cuerpo al navegar, para no abrir resultados a mitad del contenido."""
        await asyncio.sleep(0)
        await self.body.scroll_to(offset=0, duration=0)

    def _pause(self):
        """Cancela tareas de reproducción o resultados anteriores mediante revision."""
        self.playing = False
        self.revision += 1
        self.play_button.content = "Reproducir"

    def invalidate(self, e=None):
        """Marca cambios pendientes conservando resultados de la última configuración aplicada."""
        self._pause()
        self.message.color = MUTED
        self.message.value = "Cambios pendientes. Pulsa Ver resultados para aplicar la configuración."
        self.page.update()

    def population_changed(self, individuals):
        """Recibe el evento del editor; el tamaño aplicado proviene de su lista guardada."""
        self.invalidate()

    def change_mutation(self, e=None):
        """Actualiza porcentaje visible; el motor recibirá este valor dividido entre 100."""
        self.mutation_label.value = f"Mutación · {self.mutation.value:.0f} %"
        self.invalidate()

    def change_objective(self, e=None):
        """Adapta dimensiones y límites del editor; vacía individuos del esquema anterior."""
        objective = get_objective(self.function.value, int(self.dimensions.value))
        self.editor.set_schema(objective.dimensions, objective.labels, objective.bounds, notify=False)
        self.formula.value = objective.expression+"\n"+" · ".join(f"{label} ∈ [{low:g}, {high:g}]" for label, (low, high) in zip(objective.labels, objective.bounds))
        self.invalidate()
        self.message.value = "La función o las variables cambiaron. Agrega individuos o genera una nueva población."
        self.page.update()

    def configure(self, e=None):
        """Reemplaza la población por una muestra del dominio; no ejecuta el algoritmo."""
        try:
            count = int(self.size.value)
            objective = get_objective(self.function.value, int(self.dimensions.value))
            self.editor.set_individuals(initial_population(count, objective))
        except (ValueError, TypeError) as error:
            self.message.value = str(error) if isinstance(error, ValueError) and str(error).startswith("La cantidad") else "La cantidad para generar debe ser un entero entre 2 y 200."
            self.message.color = ERROR
        self.page.update()

    def settings(self) -> AlgorithmConfig:
        """Construye la entrada tipada del motor a partir de parámetros e individuos guardados."""
        try:
            total = int(self.generations.value)
        except (ValueError, TypeError):
            raise ValueError("Las generaciones deben ser un número entero.") from None
        if not 1 <= total <= 500:
            raise ValueError("Las generaciones deben estar entre 1 y 500.")
        if self.editor.editing_index is not None or any(input_field.value.strip() for input_field in self.editor.fields):
            raise ValueError("Guarda el individuo pendiente o cancela sus campos antes de ver resultados.")
        objective = get_objective(self.function.value, int(self.dimensions.value))
        population = validate_population(self.editor.individuals, objective)
        return AlgorithmConfig(population, objective, self.mutation.value/100, self.selection.value, total)

    async def apply(self, e=None):
        """Ejecuta el motor fuera del hilo de interfaz, valida la salida y abre resultados."""
        if self.busy:
            return
        try:
            config = self.settings()
        except ValueError as error:
            self.message.value, self.message.color = str(error), ERROR
            self.page.update()
            return
        self._pause()
        token = self.revision
        self.busy, self.apply_button.disabled = True, True
        self.apply_button.content = "Preparando resultados…"
        self.message.value, self.message.color = "Procesando la configuración…", MUTED
        self.page.update()
        try:
            # PUNTO DE INTEGRACIÓN: el motor inyectado recibe todos los parámetros.
            snapshots = await asyncio.to_thread(self.engine.run, config)
            validate_generations(snapshots, config)
            if token != self.revision:
                self.message.value = "La configuración cambió durante el cálculo. Vuelve a aplicar."
                return
            self.config, self.snapshots = config, snapshots
            self.total, self.index = config.generations, 0
            self.history = [min(generation.scores) for generation in snapshots]
            self.timeline.max, self.timeline.divisions = self.total, self.total
            self.message.value = "Vista de demostración: selección y mutación no modifican el recorrido." if self.is_demo else "Resultados del motor conectado."
            self.result_summary.value = f"{config.objective.name} · {config.objective.dimensions} variables · {len(config.population)} individuos · {config.selection.capitalize()} · Mutación {config.mutation_rate:.0%}"
            self.render()
            self.menu.set_enabled("results", True)
            self.show_view("results")
        except Exception as error:
            self.message.value, self.message.color = f"No se pudieron preparar los resultados: {error}", ERROR
        finally:
            self.busy, self.apply_button.disabled = False, False
            self.apply_button.content = "Ver resultados"
            self.page.update()

    def render(self):
        """Dibuja una generación recibida del motor; no calcula operadores genéticos."""
        if not self.snapshots:
            return
        generation = self.snapshots[self.index]
        objective = self.config.objective
        values, scores = generation.population, generation.scores
        best = min(range(len(values)), key=lambda i: scores[i])
        metrics = [("Generación", f"{self.index} / {self.total}"), ("Mejor evaluación", f"{scores[best]:.5g}"), ("Promedio", f"{sum(scores)/len(scores):.5g}")]
        self.metrics.controls = [ft.Container(content=ft.Column([ft.Text(label, size=12, color=MUTED), ft.Text(value, size=24, weight=ft.FontWeight.W_600, color=TEXT)], spacing=6), bgcolor="#FFFFFF", padding=20, border_radius=12, border=ft.Border.all(1, BORDER), col={"xs":12, "sm":4}) for label, value in metrics]
        self._render_function(generation, best)
        self.history_chart.set_image(plot(list(enumerate(self.history)), [(self.index, self.history[self.index])], xlabel="Generación"))
        self.table.controls = [ft.Container(content=ft.Column([ft.Row([ft.Text(f"Individuo {i+1:02d}", size=13, color=TEXT), ft.Text("Mejor" if i == best else "", size=12, color=ACCENT)]), ft.Text(" · ".join(f"{label} = {v:.5g}" for label, v in zip(objective.labels, individual)), size=12, color=MUTED), ft.Text(f"f = {scores[i]:.5g}", size=13, color=TEXT)], spacing=5), padding=12, bgcolor="#E8EEF5" if i == best else SOFT, border_radius=10) for i, individual in enumerate(values)]
        self.timeline.value, self.progress.value = self.index, self.index/self.total
        self.page.update()

    def _render_function(self, generation, best: int):
        """Actualiza sólo la gráfica de función, según dimensión y cámara seleccionadas."""
        objective = self.config.objective
        values, scores = generation.population, generation.scores
        self.chart_controls.visible = objective.dimensions == 2
        self.camera_controls.visible = objective.dimensions == 2 and self.chart_mode.value == "surface"
        if objective.dimensions == 1:
            lower, upper = objective.bounds[0]
            curve = [(lower+(upper-lower)*i/300, 0) for i in range(301)]
            chart = plot([(x, objective.evaluate((x,))) for x, _ in curve], [(v[0], score) for v, score in zip(values, scores)])
        elif self.chart_mode.value == "surface":
            chart = surface3d(objective, values, best, self.elevation.value, self.azimuth.value)
        else:
            chart = heatmap(objective, values, best)
        self.function_chart.set_image(chart)

    def change_chart(self, e=None):
        """Redibuja al elegir modo o soltar el slider de cámara; conserva generación y parámetros."""
        if not self.snapshots:
            return
        generation = self.snapshots[self.index]
        best = min(range(len(generation.scores)), key=lambda i: generation.scores[i])
        self._render_function(generation, best)
        self.page.update()

    def seek(self, e=None):
        """Pausa y salta a la generación indicada por el slider."""
        self._pause()
        self.index = round(self.timeline.value)
        self.render()

    def step(self, delta: int):
        """Pausa y avanza delta generaciones, acotando el índice al recorrido."""
        self._pause()
        self.index = max(0, min(self.total, self.index+delta))
        self.render()

    def reset(self, e=None):
        """Vuelve a la generación inicial manteniendo resultados aplicados."""
        self.step(-self.index)

    async def toggle_play(self, e=None):
        """Recorre snapshots con esperas asíncronas; revision cancela una tarea anterior."""
        if not self.snapshots:
            return
        if self.playing:
            self.step(0)
            return
        if self.index == self.total:
            self.index = 0
        self.playing = True
        self.revision += 1
        token = self.revision
        self.play_button.content = "Pausar"
        self.render()
        while self.playing and token == self.revision and self.index < self.total:
            await asyncio.sleep(0.35)
            if not self.playing or token != self.revision:
                return
            self.index += 1
            self.render()
        if token == self.revision:
            self.playing = False
            self.play_button.content = "Reproducir"
            self.page.update()
````


Parámetros: self, route, can_go_back. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Menú reutilizable con indicador de vista activa y acción de volver."""
from dataclasses import dataclass
from typing import Callable
import flet as ft
from views.theme import ACCENT, BORDER, MUTED, SOFT


@dataclass(frozen=True)
class MenuItem:
    """route identifica la vista; label e icon definen su representación."""
    route: str
    label: str
    icon: ft.IconData


class Menu(ft.Container):
    """Presenta opciones; delega navegación sin conocer datos del algoritmo."""

    def __init__(self, items: list[MenuItem], on_navigate: Callable[[str], object], on_back: Callable[[], object]):
        """items define opciones; callbacks reciben ruta elegida o solicitud de volver."""
        super().__init__(padding=12, bgcolor=SOFT, border=ft.Border.all(1, BORDER), border_radius=12)
        self.buttons = {}
        self.back_button = ft.IconButton(icon=ft.Icons.ARROW_BACK, tooltip="Volver a la vista anterior", disabled=True, on_click=lambda e: on_back())
        for item in items:
            if item.route in self.buttons:
                raise ValueError(f"Opción duplicada: {item.route}")
            self.buttons[item.route] = ft.Button(item.label, icon=item.icon, elevation=0, on_click=lambda e, route=item.route: on_navigate(route))
        self.content = ft.Row([self.back_button, *self.buttons.values()], spacing=10, wrap=True)

    def set_enabled(self, route: str, enabled: bool):
        """Habilita o bloquea una opción; el router también debe validar su acceso."""
        self.buttons[route].disabled = not enabled

    def select(self, route: str, can_go_back: bool = False):
        """Destaca route sin deshabilitarla; actualiza disponibilidad del botón Volver."""
        if route not in self.buttons:
            raise ValueError(f"Opción desconocida: {route}")
        for name, button in self.buttons.items():
            button.bgcolor = ACCENT if name == route else SOFT
            button.color = "#FFFFFF" if name == route else MUTED
        self.back_button.disabled = not can_go_back
````

Parámetros: self. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Editor reutilizable de individuos vectoriales con una lista de altura acotada."""
import math
from typing import Callable
import flet as ft
from views.theme import ACCENT, BORDER, ERROR, MUTED, SOFT, TEXT, field


class PopulationEditor(ft.Container):
    """Contenedor compuesto configurable por dimensiones, etiquetas y límites opcionales."""

    def __init__(self, dimensions: int = 2, labels=None, bounds=None, on_change: Callable | None = None, on_refresh: Callable | None = None, list_height: int = 290, max_individuals: int = 200):
        """dimensions: coordenadas; labels: nombres; bounds: dominios; callbacks: cambios/refresco.

        list_height limita la lista; max_individuals limita cantidad. on_change recibe
        una copia de los vectores tras modificar la lista; on_refresh solicita repintado.
        """
        super().__init__(padding=0)
        self._on_population_change = on_change
        self._refresh_callback = on_refresh
        self.max_individuals = max_individuals
        self._individuals = []
        self.editing_index = None
        self.fields = []
        self.inputs = ft.ResponsiveRow(spacing=14, run_spacing=10)
        self.counter = ft.Text(size=13, color=MUTED)
        self.error = ft.Text(color=ERROR, size=13, visible=False)
        self.mode = ft.Text("Nuevo individuo", size=14, weight=ft.FontWeight.W_600, color=TEXT)
        self.save_button = ft.Button("Agregar individuo", icon=ft.Icons.ADD, on_click=self.save)
        self.cancel_button = ft.TextButton("Cancelar edición", on_click=self.cancel_edit, visible=False)
        self.items = ft.ListView(height=list_height, spacing=8, build_controls_on_demand=True)
        self.content = ft.Column([self.mode, self.inputs, ft.Row([self.save_button, self.cancel_button], wrap=True), self.error, ft.Divider(color=BORDER), self.counter, self.items], spacing=12)
        self.set_schema(dimensions, labels, bounds, notify=False)

    @property
    def individuals(self) -> list[tuple[float, ...]]:
        """Devuelve una copia para impedir modificaciones externas del estado interno."""
        return list(self._individuals)

    def set_schema(self, dimensions: int, labels=None, bounds=None, notify: bool = True):
        """Configura coordenadas y vacía la lista para no reinterpretar datos existentes."""
        if not isinstance(dimensions, int) or dimensions < 1:
            raise ValueError("dimensions debe ser un entero positivo.")
        labels = tuple(labels) if labels is not None else tuple(f"x{i+1}" for i in range(dimensions))
        bounds = tuple(bounds) if bounds is not None else (None,)*dimensions
        if len(labels) != dimensions or len(bounds) != dimensions:
            raise ValueError("Debe existir una etiqueta y un límite por dimensión.")
        if any(not isinstance(label, str) or not label.strip() for label in labels):
            raise ValueError("Cada etiqueta debe ser un texto no vacío.")
        for bound in bounds:
            if bound is not None and (len(bound) != 2 or not all(math.isfinite(v) for v in bound) or bound[0] >= bound[1]):
                raise ValueError("Cada límite requiere dos valores finitos en orden ascendente.")
        self.dimensions, self.labels, self.bounds = dimensions, labels, bounds
        self.fields = [field("Valor", hint_text=f"{bound[0]:g} a {bound[1]:g}" if bound else "Número real", expand=True, on_submit=self.save) for bound in bounds]
        self.inputs.controls = [ft.Container(content=ft.Row([ft.Container(content=ft.Text(label, size=14, color=ACCENT, weight=ft.FontWeight.W_600), width=38), input_field], spacing=10), col={"xs":12, "md":6}) for label, input_field in zip(labels, self.fields)]
        self._individuals = []
        self._clear_form()
        self._render_list()
        if notify:
            self._emit_population_change()

    def _validate(self, values) -> tuple[float, ...]:
        """Valida dimensión, conversión, finitud y cotas sin alterar la lista."""
        if len(values) != self.dimensions:
            raise ValueError(f"Se requieren {self.dimensions} coordenadas.")
        result = []
        for label, raw, bound in zip(self.labels, values, self.bounds):
            try:
                value = float(raw)
            except (TypeError, ValueError):
                raise ValueError(f"Introduce un número para {label} (usa punto decimal).") from None
            if not math.isfinite(value):
                raise ValueError(f"{label} debe ser un número finito.")
            if bound is not None and not bound[0] <= value <= bound[1]:
                raise ValueError(f"{label} debe estar entre {bound[0]:g} y {bound[1]:g}.")
            result.append(value)
        return tuple(result)

    def set_individuals(self, individuals, notify: bool = True):
        """Sustituye la lista validada de forma atómica; permite lista vacía."""
        if len(individuals) > self.max_individuals:
            raise ValueError(f"Máximo {self.max_individuals} individuos.")
        values = [self._validate(individual) for individual in individuals]
        self._individuals = values
        self._clear_form()
        self._render_list()
        if notify:
            self._emit_population_change()

    def save(self, e=None):
        """Añade o sustituye el individuo editado y muestra errores junto al formulario."""
        try:
            individual = self._validate([input_field.value for input_field in self.fields])
            if self.editing_index is None:
                if len(self._individuals) >= self.max_individuals:
                    raise ValueError(f"Máximo {self.max_individuals} individuos.")
                self._individuals.append(individual)
            else:
                self._individuals[self.editing_index] = individual
        except ValueError as error:
            self.error.value, self.error.visible = str(error), True
            self._refresh()
            return
        self._clear_form()
        self._render_list()
        self._emit_population_change()

    def edit(self, index: int):
        """Carga el individuo index en los campos y activa modo edición."""
        self.editing_index = index
        for input_field, value in zip(self.fields, self._individuals[index]):
            input_field.value = repr(value)
        self.mode.value = f"Editar individuo {index+1:02d}"
        self.save_button.content = "Guardar cambios"
        self.cancel_button.visible = True
        self.error.visible = False
        self._render_list()
        self._refresh()

    def delete(self, index: int):
        """Elimina index y reajusta la edición si se elimina una fila anterior."""
        self._individuals.pop(index)
        if self.editing_index == index:
            self._clear_form()
        elif self.editing_index is not None and index < self.editing_index:
            self.editing_index -= 1
            self.mode.value = f"Editar individuo {self.editing_index+1:02d}"
        self._render_list()
        self._emit_population_change()

    def cancel_edit(self, e=None):
        """Descarta los campos pendientes sin cambiar los individuos guardados."""
        self._clear_form()
        self._render_list()
        self._refresh()

    def _clear_form(self):
        """Restablece campos, mensajes y botones para agregar un individuo nuevo."""
        self.editing_index = None
        for input_field in self.fields:
            input_field.value = ""
        self.mode.value = "Nuevo individuo"
        self.save_button.content = "Agregar individuo"
        self.cancel_button.visible = False
        self.error.visible = False

    def _render_list(self):
        """Construye filas con callbacks que capturan el índice individual."""
        self.counter.value = f"{len(self._individuals)} individuos · {self.dimensions} {'variable' if self.dimensions == 1 else 'variables'}"
        self.items.controls = []
        if not self._individuals:
            self.items.controls = [ft.Container(content=ft.Text("Aún no hay individuos. Agrega uno o genera una población.", color=MUTED), padding=18, bgcolor=SOFT, border_radius=10)]
        for index, individual in enumerate(self._individuals):
            coordinates = "    ·    ".join(f"{label} = {value:.5g}" for label, value in zip(self.labels, individual))
            self.items.controls.append(ft.Container(content=ft.Row([
                ft.Text(f"{index+1:02d}", size=12, color=MUTED, width=28),
                ft.Text(coordinates, size=14, color=TEXT, expand=True),
                ft.IconButton(icon=ft.Icons.EDIT_OUTLINED, tooltip=f"Editar individuo {index+1}", icon_color=ACCENT, on_click=lambda e, i=index: self.edit(i)),
                ft.IconButton(icon=ft.Icons.DELETE_OUTLINE, tooltip=f"Eliminar individuo {index+1}", icon_color=MUTED, on_click=lambda e, i=index: self.delete(i)),
            ], spacing=8), padding=8, bgcolor="#E9EFF6" if index == self.editing_index else SOFT, border_radius=10))

    def _emit_population_change(self):
        """Entrega una copia de los vectores al consumidor y solicita repintado."""
        if self._on_population_change:
            self._on_population_change(self.individuals)
        self._refresh()

    def _refresh(self):
        """Actualiza por callback o por la página si el control ya está montado."""
        if self._refresh_callback:
            self._refresh_callback()
        else:
            try:
                self.page
            except RuntimeError:
                # Antes de montar el componente sólo cambia el estado en memoria.
                return
            self.update()
````

### `_emit_change`
```python

```
Notifica al consumidor para sincronizar menú, reproducción y página.

Parámetros: self. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Navegación local entre vistas persistentes, con historial y condiciones de acceso."""
from typing import Callable
import flet as ft


class ViewRouter:
    """Mantiene las instancias de las vistas para conservar formularios y resultados."""

    def __init__(self, routes: dict[str, ft.Control], initial: str, guards: dict[str, Callable[[], bool]] | None = None, on_change: Callable[[str], None] | None = None):
        """routes asocia nombres y controles; guards limita acceso; on_change notifica navegación."""
        if initial not in routes:
            raise ValueError(f"Ruta inicial desconocida: {initial}")
        self.routes = dict(routes)
        self.guards = guards or {}
        self.on_change = on_change
        self.active = initial
        self.history = []
        self.outlet = ft.Column(list(self.routes.values()), spacing=0)
        self._sync_visibility()

    @property
    def can_go_back(self) -> bool:
        """Indica si existe una vista anterior a la que esté permitido volver."""
        return any(self._allowed(route) for route in self.history)

    def _allowed(self, route: str) -> bool:
        """Comprueba que la ruta exista y cumpla su condición de acceso opcional."""
        if route not in self.routes:
            raise ValueError(f"Ruta desconocida: {route}")
        return route not in self.guards or bool(self.guards[route]())

    def go(self, route: str) -> bool:
        """Abre route y registra la vista anterior; devuelve False si se bloqueó el acceso."""
        if not self._allowed(route):
            return False
        if route == self.active:
            return True
        self.history.append(self.active)
        self.active = route
        self._sync_visibility()
        self._emit_change()
        return True

    def back(self) -> bool:
        """Vuelve a la última ruta permitida, sin añadir una entrada nueva al historial."""
        while self.history:
            route = self.history.pop()
            if self._allowed(route):
                self.active = route
                self._sync_visibility()
                self._emit_change()
                return True
        return False

    def _sync_visibility(self):
        """Muestra sólo la vista activa sin recrear sus controles ni valores."""
        for name, view in self.routes.items():
            view.visible = name == self.active

    def _emit_change(self):
        """Notifica al consumidor para sincronizar menú, reproducción y página."""
        if self.on_change:
            self.on_change(self.active)
````

````python
"""Paleta y componentes compartidos de la interfaz."""
import flet as ft

BACKGROUND = "#F4F5F7"
SURFACE = "#FFFFFF"
SOFT = "#F0F3F7"
TEXT = "#243247"
MUTED = "#58677C"
ACCENT = "#456A91"
BORDER = "#DDE3EB"
ERROR = "#A33737"


def field(label: str, value: str = "", **kwargs) -> ft.TextField:
    """Campo uniforme; kwargs permite pasar eventos, anchura y otras opciones de Flet."""
    return ft.TextField(label=label, value=value, text_size=14, filled=True, fill_color=SOFT, border=ft.OutlineInputBorder(border_radius=10, side=ft.BorderSide(color=BORDER)), **kwargs)


def card(title: str, subtitle: str, controls: list, col=12) -> ft.Container:
    """Superficie clara con encabezado; col determina el ancho en ResponsiveRow."""
    return ft.Container(content=ft.Column([ft.Text(title, size=19, weight=ft.FontWeight.W_600, color=TEXT), ft.Text(subtitle, size=13, color=MUTED), *controls], spacing=18), bgcolor=SURFACE, border=ft.Border.all(1, BORDER), padding=24, border_radius=16, col=col)
````

````python
"""Genera diez notebooks enlazados y una referencia del código real de la app."""
import ast
import inspect
import json
from pathlib import Path
import flet as ft
from demo_engine_lesson import build_lesson

ROOT = Path(__file__).resolve().parents[1]
DEST = ROOT / "documentation_staging" / "AM"
NAMES = ["00_indice", "01_python_y_flet", "02_modelos", "03_controles", "04_graficas", "05_eventos", "06_referencia", "07_validacion_e_integracion", "08_conectar_algoritmo", "09_demo_engine_paso_a_paso"]


def md(text: str) -> dict:
    """Convierte text en celda Markdown conservando saltos de línea."""
    return {"cell_type": "markdown", "metadata": {}, "source": text.splitlines(keepends=True)}


def code(text: str) -> dict:
    """Convierte text en celda Python sin resultados ni contador de ejecución."""
    return {"cell_type": "code", "metadata": {}, "source": text.splitlines(keepends=True), "outputs": [], "execution_count": None}


def save(index: int, title: str, cells: list):
    """Escribe capítulo index con title, navegación a todos los notebooks e IDs estables."""
    navigation = " · ".join(f"[{name}]({name}.ipynb)" for name in NAMES)
    cells.insert(0, md(f"# {title}\n\n{navigation}\n\n**AM / Evolución Lab · Flet 1.0.4 · Python 3**"))
    for i, cell in enumerate(cells):
        cell["id"] = f"chapter-{index}-cell-{i}"
    notebook = {"nbformat":4, "nbformat_minor":5, "metadata":{"kernelspec":{"display_name":"Python 3 (AM)", "language":"python", "name":"python3"}, "language_info":{"name":"python"}}, "cells":cells}
    DEST.mkdir(parents=True, exist_ok=True)
    (DEST / f"{NAMES[index]}.ipynb").write_text(json.dumps(notebook, indent=2, ensure_ascii=False), encoding="utf-8")


save(0, "Aprender Flet desde cero", [md("""## Objetivo y recorrido
La app tiene dos vistas: Configuración y Resultados. Usa tema claro, editor de individuos por coordenadas y funciones de una o dos variables. El motor por defecto es ilustrativo, sin selección, cruce ni mutación real.

1. **Python y Flet**: entorno, argumentos, objetos, página y primer callback.
2. **Modelos**: vectores, objetivos, configuración y generaciones.
3. **Controles**: editor compuesto, propiedades y diseño adaptable.
4. **Gráficas**: curva 1D, mapa de calor 2D y transformaciones.
5. **Eventos**: estado, ejecución en hilo y reproducción.
6. **Referencia**: módulos, clases, métodos, parámetros y argumentos importados.
7. **Validación**: ejercicios ejecutables y revisión manual.
8. **Conectar tu algoritmo**: archivos a cambiar y contrato de integración.
9. **DemoEngine paso a paso**: cada apartado, comprensiones anidadas, generadores, repetición de secuencias y empaquetado/desempaquetado con bucles equivalentes ejecutables.

## Preparación
Selecciona el intérprete `D:\\Storage\\venvs\\AM\\Scripts\\python.exe` en VS Code. Puedes leer todos los notebooks sin ejecutarlos. Para ejecutar celdas necesitas Jupyter e ipykernel en el kernel elegido; no son necesarios para la app. Activa Activate.ps1 y, desde AM, ejecuta `python src/main.py`.

## Vocabulario
Módulo: archivo importable. Clase: definición de objetos. Instancia: objeto concreto. Método: función de una clase. Parámetro: nombre en una definición. Argumento: valor pasado al llamarla. Callback: función asociada a un evento. Dimensión: una coordenada del individuo. En esta app 2D significa dos entradas x,y; f(x,y) se representa por color.

La referencia documenta todos los símbolos y argumentos utilizados por este proyecto; no reproduce toda la API de Flet. Las firmas de dependencias se obtienen del entorno instalado.
""")])

save(1, "Python y Flet desde cero", [md("""## Importar y crear objetos
`import flet as ft` carga el módulo y crea alias. `from views.population_editor import PopulationEditor` importa una clase local. Python incluye src en la búsqueda cuando ejecutas src/main.py.

`ft.Text('Hola', size=24)` crea una instancia; 'Hola' es argumento posicional y size argumento nombrado. Una clase derivada, como PopulationEditor(ft.Container), hereda comportamiento del contenedor. `super().__init__` inicializa la base. `self` identifica la instancia. `@property` expone un dato mediante acceso como editor.individuals.

`tuple[float,...]` es un vector de longitud variable; list almacena elementos editables; tuple es inmutable. Las anotaciones no validan automáticamente entradas. dataclass genera constructor a partir de campos; frozen=True impide reasignarlos. Protocol define métodos esperados sin obligar a heredar de él.

## Página y callbacks
ft.run(main) inicia la app y entrega Page. page.add monta controles; cambiar una propiedad modifica el objeto; page.update comunica cambios. El guard __name__ evita iniciar la app al importar módulos.

on_click recibe una función sin ejecutarla; saludar() ejecutaría inmediatamente. e es el evento, aunque no siempre se necesita leerlo. Las lambdas de filas guardan i=index para que cada botón recuerde su propio individuo.

## Ejemplo opcional
La siguiente celda abre una ventana sólo si descomentas ft.run. Ciérrala para terminar.
"""), code("import flet as ft\n\ndef ejemplo(page: ft.Page):\n    texto = ft.Text('Hola, Flet', size=24)\n    def saludar(e):\n        texto.value = 'Has pulsado el botón'\n        page.update()\n    page.add(ft.Column([texto, ft.Button('Saludar', on_click=saludar)]))\n\n# ft.run(ejemplo)"), md("Text.value es texto; size tamaño; Column.controls lista vertical; Button.content etiqueta; on_click acción. try/except captura errores y raise ValueError los comunica. async def crea coroutine; await cede ejecución. Consulta [Flet](https://flet.dev/docs/).")])

save(2, "Modelos y contratos vectoriales", [md("""## Archivos
contracts.py define datos y validación sin Flet; objectives.py registra funciones; demo_engine.py genera una población y un recorrido ilustrativo. Antes estas tareas estaban reunidas en demo.py; ahora están separadas para permitir reemplazar el motor.

## Tipos y parámetros
Individual = tuple[float,...]. En 1D un individuo es `(x,)`, con coma para crear tupla; en 2D es `(x,y)`.

Objective(name,expression,labels,bounds,evaluate): nombre, fórmula, etiqueta por coordenada, pares (mínimo,máximo) y callable que recibe el vector completo. dimensions devuelve len(labels).

AlgorithmConfig(population,objective,mutation_rate,selection,generations): población inmutable aplicada, objetivo, probabilidad [0,1], clave torneo/ruleta/elitista y número de iteraciones. La población determina su tamaño; el campo Cantidad para generar sólo configura la muestra automática.

Generation(index,population,scores): índice consecutivo desde 0, vectores y puntuación por vector. EvolutionEngine.run(config) devuelve una lista de estas generaciones.

## Funciones provisionales
| Clave | Dominio por variable | Evaluación |
|---|---|---|
| sphere | −5 a 5 | suma de cuadrados |
| rastrigin | −5.12 a 5.12 | 10d + suma(v²−10cos(2πv)) |
| ackley | −5 a 5 | fórmula de Ackley con medias de cuadrados y cosenos |
| quartic | −2 a 4 | suma(v⁴−3v³+2) |
| wave | −6 a 6 | suma(sin(3v)+0.1v²) |

get_objective(key,dimensions) crea versión de 1 o 2 variables con etiquetas x,y. Todos los ejemplos se minimizan. math provee exp/sin/cos/pi/e, y typing Callable expresa la firma evaluadora.

## Validación y demostración
validate_population exige 2–200 individuos, cantidad correcta de coordenadas, números finitos y dominio válido. validate_generations exige generación 0 y todas las iteraciones, índices consecutivos, tamaño constante y scores finitos.

initial_population(size,objective) usa random.Random(42) y uniform por coordenada; es reproducible. DemoEngine.run muestrea 61 puntos por eje, combina con itertools.product y busca el mínimo aproximado. Interpola cada individuo hacia esa posición. No garantiza mejora monótona y no utiliza selección ni mutación.
"""), code("from pathlib import Path\nimport sys\ncandidates = [Path.cwd(), *Path.cwd().parents]\nproject = next((base/'AM' for base in candidates if (base/'AM'/'src').is_dir()), None)\nif project is None:\n    project = next(base for base in candidates if (base/'src').is_dir())\nsys.path.insert(0, str(project/'src'))\nfrom models.objectives import get_objective\nfrom models.demo_engine import initial_population, DemoEngine\nfrom models.contracts import AlgorithmConfig, validate_population, validate_generations\nobjective = get_objective('sphere', 2)\npopulation = validate_population(initial_population(5, objective), objective)\nconfig = AlgorithmConfig(population, objective, 0.05, 'torneo', 3)\nresult = DemoEngine().run(config)\nvalidate_generations(result, config)\nassert result[0].population == population\nprint(result[-1])"), md("Path.cwd obtiene carpeta del kernel; parents sus ancestros; is_dir verifica; next busca el primer candidato; sys.path.insert prioriza src. Si el kernel inicia lejos del proyecto, establece project manualmente. Esta celda prepara imports para el capítulo 07.")])

save(3, "Editor compuesto y diseño", [md("""## Dos vistas
Configuración agrupa parámetros y población. Resultados contiene métricas, gráfica, individuos actuales y recorrido. visible alterna vistas; volver a Configuración pausa reproducción. Fondo gris suave, tarjetas blancas y acento azul se definen en theme.py. ThemeMode.LIGHT activa tema claro; Theme(color_scheme_seed=ACCENT) armoniza controles Material.

## Menú y router
Menu en views/menu.py es un Container con opciones y flecha Volver. MenuItem(route,label,icon) identifica cada opción. Menu(items,on_navigate,on_back) recibe callbacks; on_navigate recibe ruta y on_back no recibe argumentos. set_enabled(route,enabled) bloquea una opción; select(route,can_go_back=False) destaca la activa y actualiza Volver. buttons conserva botones por ruta; back_button es la flecha.

ViewRouter(routes,initial,guards=None,on_change=None) en views/router.py mantiene instancias de controles. routes asocia nombres a vistas; initial es ruta inicial; guards asocia rutas con callbacks que devuelven bool; on_change recibe ruta activa. outlet aloja vistas; active guarda la ruta y history las anteriores. go(route) verifica acceso y cambia visible; ir a la misma ruta no añade historial. back() vuelve a la última ruta permitida; can_go_back consulta disponibilidad. _allowed valida acceso; _sync_visibility alterna visible; _emit_change notifica.

Es navegación local de escritorio, sin modificar una URL. Resultados se bloquea hasta aplicar una configuración válida. Cambiar de vista conserva formulario, edición pendiente e historial calculado. _on_route_change pausa reproducción y sincroniza menú. Añade nuevas vistas registrando un control en routes y un MenuItem con la misma ruta.

## PopulationEditor
Es una clase derivada de ft.Container. Su contenido combina Column, ResponsiveRow, campos, botones y ListView. La etiqueta de cada coordenada aparece a la izquierda y el campo a la derecha.

| Parámetro | Valor predeterminado | Papel |
|---|---|---|
| dimensions | 2 | cantidad de coordenadas, entero positivo |
| labels | None | nombres; por defecto x1,x2,... |
| bounds | None | límites opcionales por coordenada; None permite cualquier real finito |
| on_change | None | callback que recibe copia de población tras cambios guardados |
| on_refresh | None | callback sin argumentos para solicitar repintado |
| list_height | 290 | altura de la lista desplazable |
| max_individuals | 200 | cantidad máxima |

El editor puede tener tres o más dimensiones. La app y las gráficas de ejemplo permiten sólo una o dos. labels y bounds deben tener dimensions elementos. Los límites son pares finitos y ascendentes.

## Métodos del editor
set_schema(dimensions,labels,bounds,notify=True) redefine campos y vacía población para evitar reinterpretar coordenadas. set_individuals(individuals,notify=True) reemplaza todos los valores después de validarlos; si falla, conserva la lista anterior. individuals devuelve una copia, no una referencia mutable interna.

save(e=None) agrega o modifica según editing_index. edit(index) carga valores exactos con repr, aunque la lista muestre cinco cifras significativas. delete(index) borra y reajusta una edición activa si se elimina una fila anterior. cancel_edit(e=None) descarta el formulario pendiente. on_submit de cada campo usa save; Enter equivale al botón.

_validate(values) convierte a float y comprueba coordenadas; _clear_form restablece campos; _render_list construye filas; _emit_population_change notifica; _refresh actualiza por callback o si está montado. Evitamos llamar _notify a nuestro método porque Flet usa ese nombre internamente.

## Controles y argumentos
Text(value,size,color,weight): contenido, tamaño, tinte y grosor. TextField(label,value,text_size,filled,fill_color,border,hint_text,expand,on_submit,on_change): entrada y estilo. border recibe OutlineInputBorder(border_radius=10,side=BorderSide(color=BORDER)); redondeo y color se definen en el objeto de borde. Se evita TextField.border_color y TextField.border_radius, obsoletos desde 1.0.0. Dropdown(label,value,options,on_select): selección única. DropdownOption(key,text): clave y etiqueta. ListView(height,spacing,build_controls_on_demand): lista acotada. Button(content,icon,on_click,elevation,bgcolor,color) define acción, relieve, fondo y texto. IconButton(icon,tooltip,icon_color,on_click): acciones por fila.

Container(content,padding,bgcolor,border,border_radius,col,width): estructura y estilo. Border.all(width,color): contorno. ResponsiveRow(spacing,run_spacing) divide en 12 columnas; col={'xs':12,'lg':4} cambia ancho por tamaño de pantalla. Row(wrap=True) permite partir la fila. Column(scroll=AUTO,expand=True) da desplazamiento a la página.

field(label,value='',**kwargs) aplica estilo compartido y pasa kwargs a TextField. card(title,subtitle,controls,col=12) construye tarjetas. No pases en kwargs una opción ya fijada por field sin adaptar ese helper.

[ListView](https://flet.dev/docs/controls/listview/) · [Container](https://flet.dev/docs/controls/container/)
"""), code("# Ejemplo de componente reutilizable; no monta una ventana.\nfrom views.population_editor import PopulationEditor\neditor = PopulationEditor(dimensions=3, labels=('Ancho','Alto','Peso'), bounds=((0,10),(0,10),(0,100)))\neditor.set_individuals([(1,2,30)], notify=False)\nprint(editor.individuals)")])

save(4, "Gráficas 1D y 2D", [md("""## Curvas
plot(curve,dots,color=ACCENT,xlabel='x') recibe curva no vacía y puntos (x,y), y devuelve bytes PNG. xlabel nombra el eje horizontal; el historial utiliza 'Generación'. Matplotlib dibuja el gráfico en memoria usando Agg, sin abrir ventanas auxiliares ni usar pyplot.

_figure(height=320) crea Figure(figsize=(8,height/100),dpi=100,layout='constrained',facecolor='white') y axes con add_subplot(111). figsize mide pulgadas; dpi resolución; constrained reserva espacio para etiquetas. tick_params ajusta colores/tamaño; spines son contornos. axes.plot dibuja curva y scatter puntos; grid rejilla; set_xlabel/set_ylabel nombran ejes.

_png(figure) crea BytesIO, renderiza con FigureCanvasAgg(figure).print_png(stream), obtiene stream.getvalue y limpia figure.clear. BytesIO es un archivo binario en memoria. Esto reemplaza cientos de primitivas Canvas por una propiedad src de imagen, reduciendo el trabajo de parches de Flet.

## Mapa de calor
heatmap(objective,population,best) requiere dos variables y devuelve PNG. Muestrea una malla 40×40. LinearSegmentedColormap.from_list(name,colors) crea la paleta clara/azul. imshow(values,extent=(xmin,xmax,ymin,ymax),origin='lower',aspect='auto',cmap=palette) transforma la matriz en mapa; extent fija dominio y origin orientación. scatter añade individuos y el mejor en terracota. figure.colorbar(surface,ax=axes,pad=0.025) explica los colores. set_label y tick_params configuran leyenda. Oscuro significa mayor valor, no mejor en minimización.

La entrada 2D es (x,y), no (x,f(x)): la evaluación se representa por color. Si necesitas una superficie 3D rotatoria, añade otra visualización, conservando el contrato del motor. No es necesaria para editar ni recorrer individuos 2D.

## Chart persistente
## Superficie tridimensional
Las funciones con dos entradas admiten superficie 3D como opción predeterminada. surface3d(objective,population,best,elevation=30,azimuth=-55) dibuja z=f(x,y) y devuelve PNG. elevation y azimuth son grados de cámara; el mejor individuo aparece en terracota. NumPy.linspace(start,stop,num) muestrea 55 posiciones por eje, meshgrid crea matrices x/y y array convierte evaluaciones a matriz z.

Figure.add_subplot(111,projection='3d',computed_zorder=False) crea ejes tridimensionales. plot_surface(x,y,z,cmap,alpha,linewidth,antialiased,rcount,ccount,zorder) define superficie: alpha transparencia, rcount/ccount resolución y zorder orden de dibujo. scatter(x,y,z,depthshade=False,...) añade individuos; sus alturas son evaluaciones, no genes adicionales. view_init(elev,azim) orienta cámara y set_box_aspect((1,1,0.65)) controla proporciones. Los marcadores se dibujan por delante para distinguir la población, incluso cuando el relieve la taparía.

Dashboard.chart_mode elige surface/heatmap; chart_controls se muestra sólo en dos variables y camera_controls sólo en superficie. azimuth y elevation son sliders con on_change_end: se redibuja al soltarlos para evitar cálculos continuos. change_chart(e=None) conserva generación y llama _render_function(generation,best), sin recalcular el motor. El zoom/arrastre del visor transforma la imagen; la cámara se cambia con los sliders, no arrastrando la superficie.

## Chart persistente
Chart(height=320) hereda Container. Su ancho ocupa el disponible y su alto es explícito. raster es Image(src=bytes,width=800,height=height,fit=CONTAIN,gapless_playback=True): src contiene PNG; width/height proporcionan tamaño inicial, fit mantiene proporción y gapless_playback conserva imagen anterior mientras se carga la nueva. Una imagen mínima inicial evita contenido invisible, porque InteractiveViewer exige content visible.

viewer es InteractiveViewer(content=raster,constrained=True,min_scale=1,max_scale=4,trackpad_scroll_causes_scale=True,height=height,width=infinito). constrained aplica las restricciones disponibles al hijo. set_image(png) cambia src conservando visor y zoom. No utiliza callbacks on_resize que reconstruyan dibujos. Los valores exactos están en la lista de resultados; no hay tooltips de puntos.

La gráfica de historial muestra todas las generaciones recibidas y marca la visible. No significa ejecución en streaming ni garantiza mejora monótona.

[InteractiveViewer](https://flet.dev/docs/controls/interactiveviewer/)
""")])

save(5, "Eventos, estado y cálculo", [md("""## Estado
page aloja controles; engine implementa el motor; config es la última configuración aplicada; snapshots son Generation; history contiene mínimos; index es generación visible; total iteraciones; playing reproducción; busy cálculo en curso; revision token de invalidación; is_demo controla etiqueta de demostración.

Controles: function, dimensions, size, generations, selection, mutation_label, mutation, message, formula, editor, metrics, function_chart, history_chart, table, timeline, play_button, progress, apply_button, edit_button, action_hint, actions, result_summary, menu, configuration_view y results_view. router mantiene navegación e historial. body es la columna desplazable; raíz no tiene scroll, por lo que menú y barra actions permanecen visibles. _on_route_change alterna Ver resultados y Editar configuración.

Al navegar se programa _scroll_to_start mediante page.run_task: espera un turno con asyncio.sleep(0) y llama await body.scroll_to(offset=0,duration=0). Así los resultados se abren desde arriba y la gráfica no queda oculta por el desplazamiento del formulario anterior. Sólo cambia desplazamiento, conservando datos y ediciones.

## Preparación
_prepare_population crea muestra inicial sin ejecutar el motor. build organiza vistas. show_view(name) delega navegación al router; _on_route_change(name) pausa y sincroniza menú. change_objective reinicia esquema y vacía lista al cambiar dominio/dimensiones. configure genera población sin aplicarla. population_changed(individuals) marca cambios; la cantidad aplicada deriva de la lista, no del campo del generador.

invalidate detiene reproducción y marca cambios pendientes. change_mutation actualiza etiqueta; settings devuelve AlgorithmConfig con mutation.value/100. Si hay campos pendientes o edición activa, settings pide guardar o cancelar antes de aplicar.

## Ejecutar motor
apply es async: valida entradas, pausa, marca busy, deshabilita botón, captura token, ejecuta `await asyncio.to_thread(self.engine.run, config)` y valida generaciones. El hilo permite atender eventos mientras trabaja el motor. La configuración es inmutable para evitar modificar sus vectores durante el cálculo.

Si cambió revision, el resultado se descarta y la configuración aplicada anterior se conserva. Esta invalidación no cancela físicamente el hilo del motor; una cancelación real requiere cooperación del motor. finally restablece botón y busy incluso cuando hay errores.

No necesitas cambiar render para integrar operadores: render presenta una Generation ya calculada, métricas, mapa/curva, lista, historial y progreso. La minimización usa min(scores); si tu algoritmo maximiza, adapta ese criterio en render/history y documentación.

## Recorrido
seek lee slider y pausa; step(delta) limita índice; reset vuelve a 0. toggle_play usa await asyncio.sleep(0.35); token controla si otra interacción canceló el recorrido. Al terminar restaura Reproducir. _pause incrementa revision y detiene reproducciones antiguas.

e=None permite callbacks y llamadas internas; self es instancia. lambda e,i=index captura cada fila. Los cambios de datos llaman page.update después de agrupar propiedades.
""")])

cells = [md("## Alcance\nSe documentan módulos propios, clases, métodos y todas las llamadas usadas, incluidos símbolos importados. Los bloques de código real son de lectura: no ejecutes todos los bloques de esta referencia, porque algunos son puntos de entrada o herramientas que generan archivos.")]
sources = sorted((ROOT / "src").rglob("*.py"))+sorted((ROOT / "tools").glob("*.py"))
for path in sources:
    source = path.read_text(encoding="utf-8")
    tree = ast.parse(source)
    cells.append(md(f"## Módulo `{path.relative_to(ROOT)}`\n{ast.get_docstring(tree) or 'Marcador de paquete vacío; agrupa módulos sin definir comportamiento.'}"))
    imports = [ast.unparse(n) for n in ast.walk(tree) if isinstance(n,(ast.Import,ast.ImportFrom))]
    cells.append(md("### Importaciones\n```python\n"+"\n".join(imports)+"\n```"))
    for node in ast.walk(tree):
        if isinstance(node, ast.ClassDef):
            fields = [ast.unparse(n) for n in node.body if isinstance(n,ast.AnnAssign)]
            cells.append(md(f"### Clase `{node.name}`\n{ast.get_docstring(node) or 'Clase auxiliar de comprobación; no forma parte de la app.'}\n```python\n"+"\n".join(fields)+"\n```"))
        if isinstance(node,(ast.FunctionDef,ast.AsyncFunctionDef)):
            params = [a.arg for a in node.args.args]
            if node.args.kwarg:
                params.append("**"+node.args.kwarg.arg)
            cells.append(md(f"### `{node.name}`\n```python\n{ast.unparse(node).splitlines()[0]}\n```\n{ast.get_docstring(node) or 'Callback o utilidad local; su comportamiento está en el código adjunto.'}\n\nParámetros: {', '.join(params) or 'ninguno'}. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento."))
    cells.append(code(source if source.strip() else "# Paquete vacío."))
    calls = sorted(set(ast.unparse(n) for n in ast.walk(tree) if isinstance(n,ast.Call)))
    cells.append(md("### Inventario de llamadas y argumentos exactos\nPosicionales en orden y nombrados con su parámetro; las expresiones muestran procedencia de valores.\n```python\n"+"\n".join(calls)+"\n```"))

symbols = {
    "Text":"value contenido; size tamaño; weight grosor; color tinte; expand distribuye espacio; width limita ancho.",
    "TextField":"label título; value dato; text_size fuente; filled/fill_color fondo; border objeto OutlineInputBorder; hint_text guía; expand ancho; on_submit Enter; on_change edición.",
    "OutlineInputBorder":"border_radius redondea contorno; side recibe BorderSide, evitando propiedades obsoletas del campo.",
    "BorderSide":"color define tinte; width conserva grosor predeterminado.",
    "Dropdown":"label título; value clave; options lista; on_select callback.",
    "DropdownOption":"key identificador; text etiqueta.",
    "Slider":"min/max límites; divisions pasos; value posición; label plantilla; active_color color; on_change callback.",
    "ListView":"controls filas; height limita área; spacing separación; build_controls_on_demand crea filas según necesidad.",
    "Button":"content etiqueta; icon símbolo; on_click callback; disabled bloqueo.",
    "TextButton":"content etiqueta; icon opcional; on_click acción; disabled bloqueo.",
    "IconButton":"icon símbolo; tooltip ayuda; icon_color tinte; on_click acción.",
    "Icon":"icon símbolo (posicional); color tinte; size tamaño.",
    "ProgressBar":"value fracción; color avance; bgcolor fondo.",
    "Column":"controls hijos verticales; spacing separación; visible alterna vista; scroll y expand ajustan página.",
    "Row":"controls hijos horizontales; spacing separación; wrap permite varias líneas.",
    "ResponsiveRow":"controls hijos; spacing horizontal; run_spacing vertical; usa col de cada hijo.",
    "Container":"content hijo; bgcolor fondo; border contorno; padding espacio; border_radius redondeo; col distribución; width ancho.",
    "Divider":"color tinte del separador.",
    "InteractiveViewer":"content gráfico; min_scale/max_scale zoom; trackpad_scroll_causes_scale cambia escala; height alto.",
    "Image":"src bytes PNG; width/height tamaños explícitos; fit=CONTAIN conserva proporción; gapless_playback evita parpadeo; semantics_label descripción accesible.",
    "Screenshot":"content árbol a capturar; expand ocupa espacio. capture(pixel_ratio=None,delay=None) es async y devuelve PNG del cliente real.",
    "TextStyle":"size tamaño; color tinte en Canvas.",
    "Paint":"color tinte; stroke_width grosor; style contorno/relleno.",
    "Theme":"color_scheme_seed genera esquema Material.",
    "Page":"title título; theme_mode y theme apariencia; bgcolor fondo; padding espacio; window.width/height/min_width/min_height límites. add(control) monta; update comunica cambios.",
}
for name, explanation in symbols.items():
    cells.append(md(f"## Clase importada `ft.{name}`\n{explanation}\nLa firma incluye también opciones no usadas de la dependencia.\n```python\n{name}{inspect.signature(getattr(ft,name))}\n```"))
cells.append(md("""## Módulos y símbolos importados adicionales
| Módulo | Símbolos / significado / argumentos usados |
|---|---|
| dataclasses | dataclass(frozen=True) genera datos inmutables; replace(instance,field=value) crea variante en pruebas |
| typing | Callable[[Individual],float] contrato evaluador; Callable callback; Protocol contrato estructural |
| math | sin(x),cos(x),exp(x),sqrt(x),isfinite(x); pi/e constantes; radianes en trigonometría |
| random | Random(42) generador local; uniform(lower,upper) posición |
| itertools | product(*axes) producto cartesiano de coordenadas muestreadas |
| asyncio | to_thread(function,*args) corre en hilo; sleep(seconds) espera; create_task(coroutine) agenda; run(coroutine) ejecuta prueba |
| threading | Event() sincroniza pruebas; set() señala; wait(timeout=...) espera; get_ident() identifica hilo |
| types | SimpleNamespace(update=lambda:None) página mínima para pruebas de controles |
| pathlib | Path(path), resolve(), parents/parent, cwd(), is_dir(), rglob(pattern), glob(pattern), relative_to(root), read_text(encoding), write_text(text,encoding), mkdir(parents,exist_ok) |
| sys | path.insert(0,str(src)) habilita imports locales |
| ast | parse(source),walk(tree),unparse(node),get_docstring(node); tipos Import,ImportFrom,ClassDef,AnnAssign,FunctionDef,AsyncFunctionDef,Call |
| inspect | signature(object) firma del constructor instalado |
| json | dumps(object,indent=2,ensure_ascii=False) serializa; loads(text) decodifica notebook |

## Flet: constantes y métodos
ThemeMode.LIGHT tema; FontWeight.W_600 grosor medio; ScrollMode.AUTO desplazamiento; CrossAxisAlignment.STRETCH ocupa ancho; BoxFit.CONTAIN mantiene proporción; Icons.* símbolos Material. Border.all(1,color) crea borde uniforme. Control.page sólo está disponible cuando el control fue montado; acceder antes causa RuntimeError. ft.run(main) inicia la app; await page.window.close() cierra prueba. _trigger_event(name,data) es API interna usada sólo por el verificador para despachar clics con cliente real, no por la app.

## Renderización y prueba de escritorio
io.BytesIO() almacena PNG en memoria; getvalue() devuelve bytes. base64.b64decode(text) decodifica PNG mínimo de inicio. matplotlib.figure.Figure configura figura; matplotlib.backends.backend_agg.FigureCanvasAgg la renderiza; matplotlib.colors.LinearSegmentedColormap crea paleta. Las llamadas y argumentos de axes.plot/scatter/grid/labels/imshow, spines, colorbar y print_png se describen en el capítulo 04 y en el inventario exacto de charts.py. zip(*points) separa coordenadas para Matplotlib.

tools/check_desktop.py usa Screenshot para capturar cliente real en 900×700 y 760×640. screenshot(name) espera frame, llama capture con asyncio.wait_for(...,timeout=15), valida cabecera PNG y usa Path.write_bytes para guardar. traceback.format_exc() conserva error; page.on_error recopila errores del cliente. La prueba hace clics por _trigger_event, reproduce, vuelve durante reproducción y alterna vistas repetidamente. Capturas y report.json se guardan en artifacts/desktop_check; failure.txt se escribe si falla. Estas comprobaciones requieren abrir una ventana real.

## Python: operaciones usadas
float/int/str convierten; len cuenta; range produce índices; min/max límites o extremos (key especifica criterio); sum suma; round redondea; enumerate empareja índice/valor; zip empareja listas; list/tuple/set materializan colecciones; any/all evalúan condiciones; sorted ordena; next obtiene primer elemento; isinstance comprueba tipo; getattr obtiene atributo; print muestra resultados; repr conserva representación exacta.

append agrega elemento; extend agrega varios; pop(index) elimina; insert(index,value) inserta; clear vacía; splitlines(keepends=True) conserva saltos; join concatena; strip elimina blancos; startswith comprueba prefijo. ValueError(message) indica dato inválido; TypeError tipo incompatible; RuntimeError estado incompatible; Exception agrupa errores del motor; AssertionError señala prueba fallida. assert comprueba condición. F-strings :g/:.5g cifras significativas, :02d índice, :.0% porcentaje; operadores / de Path unen rutas.

exec(source,namespace) ejecuta celdas de los notebooks propios revisados; no se recomienda aplicarlo a notebooks externos sin revisar. __file__ identifica herramienta. Los campos Jupyter cell_type/metadata/source/outputs/execution_count/id describen celdas; nbformat/nbformat_minor versión 4.5; kernelspec describe kernel.

## Herramientas
md(text) crea celda Markdown; code(text) celda Python; save(index,title,cells) serializa capítulo. check_editor() comprueba CRUD; check_models() contratos; check_dashboard() integración asíncrona; check_notebooks() enlaces y ejercicios. Los métodos privados usan prefijo _ por convención; las definiciones, parámetros y llamadas están incluidos en los bloques de cada módulo.
"""))
save(6, "Referencia completa de módulos y argumentos", cells)

save(7, "Pruebas y ejercicios", [md("""## Preparación
Ejecuta la celda del capítulo 02 para localizar el proyecto. Las siguientes celdas comprueban vectores, datos inválidos y CRUD sin abrir ventana.
"""), code("from models.contracts import validate_population\nfor invalid in [[(float('nan'),0),(1,2)], [(1000,0),(1,2)], [(1,),(2,)]]:\n    try:\n        validate_population(invalid, objective)\n    except ValueError:\n        pass\n    else:\n        raise AssertionError('Se aceptó población inválida')\nprint('Validaciones correctas')"), code("from views.population_editor import PopulationEditor\neditor = PopulationEditor(2, ('x','y'), ((-5,5),(-5,5)))\neditor.set_individuals([(1,2),(3,4)])\neditor.edit(1)\neditor.fields[0].value = '-1'\neditor.save()\nassert editor.individuals == [(1,2),(-1,4)]\neditor.delete(0)\nassert editor.individuals == [(-1,4)]\nprint('Editor CRUD correcto')"), md("""## Qué se comprueba
`python tools/check_interface.py` verifica agregar/editar/eliminar/cancelar, cambio de dimensiones, copia defensiva, reemplazo atómico, cinco funciones en 1D/2D, salida del motor, recepción de selección/mutación, ejecución en hilo, pausa, campos pendientes y resultados descartados si cambió configuración.

SimpleNamespace es un sustituto mínimo que no muestra una ventana. Las pruebas de construcción no prueban apariencia ni gestos reales. assert/AssertionError detectan incumplimientos.

## Revisión visual manual
1. Ejecuta python src/main.py: debe iniciar en Configuración con tema claro.
2. Cambia a una/dos variables, genera e ingresa coordenadas; la lista no debe crecer fuera de su área.
3. Edita y elimina filas; prueba texto, NaN, límites inválidos y coordenadas pendientes.
4. Ver resultados debe mostrar curva o mapa según dimensiones.
5. Reproduce, pausa, vuelve a Configuración y aplica cambios.
6. Reduce ventana, verifica desplazamiento, ejes y zoom.

## Pruebas de navegación y deprecaciones
Las comprobaciones cubren bloqueo de Resultados sin datos, navegación por botones, Volver, edición pendiente conservada y rutas inválidas. `python -W error::DeprecationWarning tools/check_interface.py` convierte advertencias en errores para detectar usos obsoletos.

## Límites
La demostración no implementa genética; el mínimo es aproximado; puede haber aumento temporal de puntuaciones. La superficie 3D rota mediante sliders y se renderiza a imagen, sin una escena 3D manipulable directamente con el ratón. No hay persistencia, streaming ni exportación. El visor conserva zoom al cambiar imagen. Las gráficas minimizan, y el contrato conserva tamaño de población. Invalidar cálculo no detiene el hilo de un motor externo.
""")])

save(8, "Dónde conectar tu algoritmo genético", [md("""## Mapa de cambios
| Archivo / apartado | Tu trabajo |
|---|---|
| src/models/objectives.py · FUNCTIONS y get_objective | define las cinco funciones reales, etiquetas, dominios y evaluate(vector) |
| src/models/genetic_engine.py · archivo nuevo | implementa selección, cruce, mutación y loop de generaciones en run(config) |
| src/main.py · comentario CAMBIA AQUÍ | importa tu clase y reemplaza DemoEngine() por GeneticEngine() |
| src/models/contracts.py | contrato de datos; consérvalo si tu motor puede adaptarse |
| src/views/dashboard.py · apply | ya entrega parámetros y recibe resultados; normalmente no necesitas cambiarlo |
| src/views/dashboard.py · render | presenta una generación; no introduzcas operadores aquí |
| src/views/population_editor.py | captura/edita vectores; reutilizable, independiente del motor |
| src/views/charts.py | curva 1D / mapa 2D; cambia sólo si deseas otra visualización |

## Entrada exacta
config.population es tuple de individuos (tuples de floats). config.objective.evaluate(individual) calcula aptitud. objective.labels nombra genes y objective.bounds fija límites. mutation_rate ya está normalizada: 0.05 significa 5 %. selection es 'torneo', 'ruleta' o 'elitista'. generations indica iteraciones, además de la generación inicial.

Tu motor puede convertir datos a NumPy internamente, pero retorna tuples de floats en Generation. En este contrato cada generación mantiene población entre 2 y 200 y la misma cantidad inicial. Si cambias tamaño dinámicamente, adapta validate_generations y documenta ese comportamiento.

## Plantilla de integración
La siguiente plantilla es intencionalmente incompleta: indica dónde escribir tus operadores, sin inventar tu algoritmo.
"""), code("# Guardar en src/models/genetic_engine.py cuando implementes el motor.\nfrom models.contracts import AlgorithmConfig, Generation\n\nclass GeneticEngine:\n    def run(self, config: AlgorithmConfig) -> list[Generation]:\n        population = config.population\n        snapshots = [Generation(0, population, tuple(config.objective.evaluate(v) for v in population))]\n        for index in range(1, config.generations + 1):\n            # 1. Selección usando config.selection.\n            # 2. Cruce de los padres.\n            # 3. Mutación usando config.mutation_rate.\n            # 4. Respeta config.objective.bounds y el tamaño original.\n            # 5. Guarda coordenadas decodificadas, aunque uses cromosomas binarios.\n            raise NotImplementedError('Implementa los operadores de tu algoritmo aquí')\n        return snapshots\n\n# En src/main.py:\n# from models.genetic_engine import GeneticEngine\n# page.add(Dashboard(page, engine=GeneticEngine()).build())"), md("""## Salida y decisiones
Devuelve generación 0 con la población original, seguida de Generation(index,population,scores) por iteración. scores tiene una evaluación finita por individuo en el mismo orden. No devuelvas solamente el mejor: la interfaz necesita la población completa para mostrar distribución y lista.

Actualmente se minimiza. Si maximizarás, cambia min en history y best dentro de Dashboard y revisa la semántica del mapa. Para selección de ruleta con valores negativos debes definir transformación de aptitud en tu motor. Torneo, ruleta y elitismo dependen de tus decisiones de implementación, no de controles Flet.

apply ejecuta run en un hilo y no muestra generaciones hasta que termina. Para progreso real durante cálculo, añade un contrato productor/consumidor o callback con control de cancelación; la reproducción actual recorre resultados ya recibidos. No actualices controles Flet desde run: entrega datos a Dashboard.

## Cómo probar tu conexión
Inyecta un motor de prueba y comprueba que recibe mutación/selección correctas. Usa validate_generations antes de mostrar resultados; tools/check_interface.py incluye una prueba de ese intercambio. Separa las pruebas de los operadores de las pruebas visuales.
""")])
save(9, "DemoEngine: Python y conexión con Flet paso a paso", build_lesson(md, code, (ROOT / "src" / "models" / "demo_engine.py").read_text(encoding="utf-8")))
print(f"Generados {len(NAMES)} notebooks en {DEST}")
````

### `screenshot`
```python
async def screenshot(name):
```
Espera un frame y obtiene una captura del árbol visual dentro de Flet.

Parámetros: name. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Verificación con cliente Flet real: capturas, botón fijo, eventos y navegación."""
import asyncio
import json
from pathlib import Path
import sys
import traceback
import flet as ft

ROOT = Path(__file__).resolve().parents[1]
sys.path.insert(0, str(ROOT / "src"))
from views.dashboard import Dashboard
from views.theme import ACCENT, BACKGROUND

DEST = ROOT / "artifacts" / "desktop_check"


async def main(page: ft.Page):
    """Abre ventana de 900×700, verifica eventos reales y guarda capturas antes de cerrar."""
    DEST.mkdir(parents=True, exist_ok=True)
    errors = []
    page.on_error = lambda e: errors.append(str(e.data))
    page.title = "Evolución Lab · Comprobación visual"
    page.window.width, page.window.height = 900, 700
    page.theme_mode = ft.ThemeMode.LIGHT
    page.theme = ft.Theme(color_scheme_seed=ACCENT)
    page.bgcolor, page.padding = BACKGROUND, 16
    panel = Dashboard(page)
    panel.generations.value = "2"
    capture = ft.Screenshot(content=ft.Container(content=panel.build(), bgcolor=BACKGROUND, expand=True), expand=True)
    page.add(capture)

    async def screenshot(name):
        """Espera un frame y obtiene una captura del árbol visual dentro de Flet."""
        await asyncio.sleep(0.7)
        png = await asyncio.wait_for(capture.capture(), timeout=15)
        assert png.startswith(b"\x89PNG")
        (DEST / name).write_bytes(png)

    try:
        await screenshot("01_config_900x700.png")
        await panel.apply_button._trigger_event("click", None)
        assert panel.router.active == "results", panel.message.value
        assert panel.function_chart.raster.src.startswith(b"\x89PNG")
        await screenshot("02_results_2d.png")
        await panel.body.scroll_to(offset=330, duration=0)
        await screenshot("06_surface_3d.png")
        panel.azimuth.value = 45
        await panel.azimuth._trigger_event("change_end", None)
        await screenshot("07_surface_rotated.png")
        panel.chart_mode.value = "heatmap"
        await panel.chart_mode._trigger_event("select", None)
        await screenshot("08_heatmap.png")
        panel.chart_mode.value = "surface"
        await panel.chart_mode._trigger_event("select", None)
        playback = asyncio.create_task(panel.play_button._trigger_event("click", None))
        await asyncio.sleep(0.1)
        await panel.menu.back_button._trigger_event("click", None)
        await playback
        assert panel.router.active == "config"
        assert not panel.playing
        await screenshot("03_back_to_config.png")
        panel.dimensions.value = "1"
        await panel.dimensions._trigger_event("select", None)
        panel.configure()
        await panel.apply_button._trigger_event("click", None)
        assert panel.router.active == "results" and panel.config.objective.dimensions == 1
        await screenshot("04_results_1d.png")
        for _ in range(5):
            await panel.menu.buttons["config"]._trigger_event("click", None)
            await panel.menu.buttons["results"]._trigger_event("click", None)
        await panel.menu.buttons["config"]._trigger_event("click", None)
        page.window.width, page.window.height = 760, 640
        page.update()
        await screenshot("05_config_760x640.png")
        assert not errors, errors
        (DEST / "report.json").write_text(json.dumps({"ok":True, "errors":errors, "checks":["Evento Ver resultados", "Gráficas PNG 1D/2D", "Evento Volver", "Menú", "Ventanas 900x700 y 760x640"]}, ensure_ascii=False, indent=2), encoding="utf-8")
        print("Cliente real: navegación y capturas verificadas.", flush=True)
    except Exception:
        (DEST / "failure.txt").write_text(traceback.format_exc(), encoding="utf-8")
        raise
    finally:
        await page.window.close()


if __name__ == "__main__":
    ft.run(main)
````

### `run`
```python
def run(self, config):
```
Callback o utilidad local; su comportamiento está en el código adjunto.

Parámetros: self, config. Defaults y retorno anotado aparecen en la firma; sin return explícito devuelve None. self identifica la instancia, e el evento.

````python
"""Pruebas de edición multidimensional, contrato del motor, tareas y notebooks."""
import asyncio
from dataclasses import replace
import json
from pathlib import Path
import sys
import threading
from types import SimpleNamespace

ROOT = Path(__file__).resolve().parents[1]
sys.path.insert(0, str(ROOT / "src"))
from models.contracts import AlgorithmConfig, Generation, validate_generations, validate_population
from models.demo_engine import DemoEngine, initial_population
from models.objectives import FUNCTIONS, get_objective
from views.dashboard import Dashboard
from views.population_editor import PopulationEditor


def check_editor():
    """Prueba CRUD, validación atómica, copias y borrado durante edición en varias dimensiones."""
    changes = []
    editor = PopulationEditor(2, ("x", "y"), ((-5, 5), (-5, 5)), on_change=changes.append, on_refresh=lambda:None)
    for input_field, value in zip(editor.fields, ("1", "2")):
        input_field.value = value
    editor.save()
    assert editor.individuals == [(1, 2)] and changes[-1] == [(1, 2)]
    editor.fields[0].value, editor.fields[1].value = "nan", "0"
    editor.save()
    assert editor.error.visible and editor.individuals == [(1, 2)]
    editor.set_individuals([(1, 2), (3, 4), (0, 0)])
    editor.edit(2)
    editor.delete(0)
    assert editor.editing_index == 1
    editor.fields[0].value = "-1"
    editor.save()
    assert editor.individuals == [(3, 4), (-1, 0)]
    editor.edit(0)
    editor.fields[0].value = "99"
    editor.cancel_edit()
    assert editor.individuals[0] == (3, 4)
    editor.edit(1)
    editor.delete(1)
    assert editor.editing_index is None
    copy = editor.individuals
    copy.clear()
    assert len(editor.individuals) == 1
    try:
        editor.set_individuals([(1, 2), (100, 0)])
    except ValueError:
        pass
    else:
        raise AssertionError("Aceptó valores fuera del dominio")
    assert editor.individuals == [(3, 4)]
    editor.set_schema(3, ("Ancho", "Alto", "Peso"))
    editor.set_individuals([(1, 2, 3)])
    assert editor.dimensions == 3 and len(editor.fields) == 3
    editor.set_schema(1, ("x",))
    assert editor.individuals == []
    standalone = PopulationEditor()
    standalone.set_individuals([(1, 2)])
    assert standalone.individuals == [(1, 2)]
    print("Editor: agregar, editar, cancelar, eliminar, validación y 1/2/3 dimensiones correctos.")


def check_models():
    """Comprueba objetivos vectoriales y rechazos del contrato de generaciones."""
    for dimension in (1, 2):
        for key in FUNCTIONS:
            objective = get_objective(key, dimension)
            population = validate_population(initial_population(6, objective), objective)
            config = AlgorithmConfig(population, objective, 0.05, "torneo", 3)
            result = DemoEngine().run(config)
            validate_generations(result, config)
            assert result[0].population == population
            assert all(len(vector) == dimension for snapshot in result for vector in snapshot.population)
    invalid = [result[:-1], [replace(result[0], scores=(float("nan"),)*6), *result[1:]], [replace(result[0], index=4), *result[1:]]]
    for snapshots in invalid:
        try:
            validate_generations(snapshots, config)
        except ValueError:
            pass
        else:
            raise AssertionError("Aceptó un contrato inválido")
    print("Cinco objetivos en 1D/2D y validación de salida del motor correctos.")


async def check_dashboard():
    """Verifica conexión de parámetros, reproducción, errores y descarte de resultados antiguos."""
    class RecordingEngine:
        def run(self, config):
            self.config = config
            self.thread = threading.get_ident()
            return DemoEngine().run(config)

    engine = RecordingEngine()
    panel = Dashboard(SimpleNamespace(update=lambda:None), engine)
    root = panel.build()
    assert root.scroll is None and root.controls[-1] is panel.actions
    assert panel.body.expand and panel.body.scroll is not None
    assert panel.function_chart.raster.width > 0 and panel.function_chart.raster.height > 0
    assert panel.configuration_view.visible and not panel.results_view.visible
    assert panel.router.active == "config"
    assert not panel.router.go("results")
    assert panel.menu.buttons["results"].disabled
    panel.selection.value, panel.mutation.value, panel.generations.value = "ruleta", 17, "2"
    await panel.apply()
    assert engine.thread != threading.get_ident()
    assert engine.config.selection == "ruleta" and engine.config.mutation_rate == 0.17
    assert panel.results_view.visible and len(panel.snapshots) == 3
    assert panel.config.objective.dimensions == 2
    assert panel.chart_controls.visible and panel.camera_controls.visible
    surface_png = panel.function_chart.raster.src
    panel.azimuth.value = 45
    panel.change_chart()
    assert surface_png != panel.function_chart.raster.src
    panel.chart_mode.value = "heatmap"
    panel.change_chart()
    assert not panel.camera_controls.visible and panel.index == 0
    panel.chart_mode.value = "surface"
    panel.change_chart()
    assert panel.function_chart.raster.src.startswith(b"\x89PNG")
    function_viewer = panel.function_chart.viewer
    history_viewer = panel.history_chart.viewer
    panel.step(1)
    assert panel.function_chart.viewer is function_viewer
    assert panel.history_chart.viewer is history_viewer
    panel.reset()
    panel.editor.edit(0)
    panel.editor.fields[0].value = "1.25"
    current_snapshots = panel.snapshots
    assert panel.show_view("config")
    assert panel.editor.editing_index == 0 and panel.editor.fields[0].value == "1.25"
    assert panel.router.back()
    assert panel.router.active == "results" and panel.snapshots is current_snapshots
    assert panel.editor.fields[0].value == "1.25"
    assert panel.show_view("config")
    panel.editor.cancel_edit()
    assert panel.show_view("results")
    history = list(panel.router.history)
    assert panel.show_view("results") and panel.router.history == history
    try:
        panel.router.go("missing")
    except ValueError:
        pass
    else:
        raise AssertionError("Aceptó ruta inexistente")
    assert panel.router.active == "results"
    task = asyncio.create_task(panel.toggle_play())
    await asyncio.sleep(0.03)
    panel.menu.buttons["config"].on_click(None)
    await task
    assert panel.index == 0 and not panel.playing and panel.router.active == "config"
    panel.menu.buttons["results"].on_click(None)
    assert panel.router.active == "results"
    panel.generations.value = "1"
    await panel.apply()
    await panel.toggle_play()
    assert panel.index == 1 and not panel.playing
    snapshots = panel.snapshots
    panel.editor.fields[0].value = "1"
    await panel.apply()
    assert panel.snapshots is snapshots and "pendiente" in panel.message.value
    panel.editor.cancel_edit()
    panel.dimensions.value = "1"
    panel.change_objective()
    assert panel.editor.dimensions == 1 and panel.editor.individuals == []
    await panel.apply()
    assert "2 y 200" in panel.message.value
    panel.configure()
    await panel.apply()
    assert panel.config.objective.dimensions == 1

    class SlowEngine:
        def run(self, config):
            started.set()
            release.wait(timeout=5)
            return DemoEngine().run(config)

    started, release = threading.Event(), threading.Event()
    panel.engine = SlowEngine()
    task = asyncio.create_task(panel.apply())
    await asyncio.to_thread(started.wait, 2)
    old_config = panel.config
    panel.invalidate()
    release.set()
    await task
    assert panel.config is old_config and "cambió" in panel.message.value
    assert not panel.busy and not panel.apply_button.disabled
    print("Motor inyectado, parámetros, cálculo en hilo, pausa y descarte de resultados antiguos correctos.")


def check_notebooks():
    """Verifica enlaces, IDs y ejecuta ejercicios de modelos/editor de la documentación."""
    files = sorted((ROOT / "documentation_staging" / "AM").glob("*.ipynb"))
    assert len(files) == 10
    namespace = {}
    count = 0
    for path in files:
        notebook = json.loads(path.read_text(encoding="utf-8"))
        assert notebook["nbformat"] == 4
        assert len({c["id"] for c in notebook["cells"]}) == len(notebook["cells"])
        for other in files:
            assert f"({other.name})" in "".join(notebook["cells"][0]["source"])
        if path.name.startswith(("02_", "07_", "09_")):
            for cell in notebook["cells"]:
                if cell["cell_type"] == "code":
                    exec("".join(cell["source"]), namespace)
                    count += 1
    print(f"10 notebooks enlazados válidos; {count} celdas prácticas ejecutadas.")


if __name__ == "__main__":
    check_editor()
    check_models()
    asyncio.run(check_dashboard())
    check_notebooks()
````

Para una coordenada inicial v y destino t, la expresión `v+(t-v)*progress` es una interpolación:\n\n- progress=0: conserva v.\n- progress=0.5: posición a mitad de camino.\n- progress=1: alcanza t, salvo redondeos propios de floats.\n\nEl signo menos calcula desplazamiento, la multiplicación aplica fracción y la suma añade el punto inicial. No son operadores de cruce ni mutación.\n')
```python
md('## 16. Expresión condicional if/else\n`valor_si_verdadero if condición else valor_si_falso` elige una expresión. En el motor:\n\n```python\npopulation = config.population if index == 0 else ...\n```\n\nSi index==0, utiliza la población original sin recalcular coordenadas. En otro caso, calcula los nuevos vectores. Sólo se evalúa la rama elegida. `==` compara; `=` asigna.\n\nEquivale a un if/else tradicional con asignación en cada rama. No es el filtro de una comprensión: siempre devuelve un valor.\n\n## 17. La población interpolada: dos generadores y dos tuple\n```python\ntuple(tuple(v+(t-v)*progress for v,t in zip(individual,target))\n      for individual in config.population)\n```\n\n1. El generador exterior recorre cada individuo original.\n2. zip(individual,target) empareja cada coordenada actual con su destino.\n3. `for v,t` desempaqueta cada pareja.\n4. La expresión numérica calcula una coordenada nueva.\n5. tuple interior reúne coordenadas de **un individuo**.\n6. tuple exterior reúne individuos de **la población**.\n\nEn este contrato, ambos vectores tienen la misma dimensión. Como zip corta al más corto, validar dimensiones antes evita perder coordenadas silenciosamente.\n')
md('## 18. Puntuaciones y empaquetado en Generation\n```python\ntuple(objective.evaluate(individual) for individual in population)\n```\n\nEste generador recorre la población nueva y calcula una puntuación por individuo. tuple materializa puntuaciones en el mismo orden que sus vectores.\n\n`Generation(index,population,scores)` construye una instancia de la dataclass Generation. No es una tupla de tres elementos: es un objeto con atributos nombrados. El orden de argumentos posicionales corresponde al de los campos declarados en contracts.py; también puedes escribir `Generation(index=index,population=population,scores=scores)`.\n\n`result.append(generation)` añade una generación a la lista y devuelve None. No hagas `result = result.append(...)`, porque reemplazarías la lista por None. `return result` termina run y entrega todas las generaciones.\n')
md('## 19. run completo con bucles tradicionales\nLa siguiente versión realiza las mismas operaciones en el mismo orden, manteniendo la misma malla y destino. Su propósito es pedagógico; no reemplaza el archivo de la app. Compara cada apartado con la versión compacta.\n')
md('## 20. Código vigente para comparar\nEl siguiente bloque se copia automáticamente del archivo al generar la documentación. Los ejemplos anteriores lo descomponen; el código de la app no se cambia para crear este capítulo.\n\n```python\n' + source + "\n```\n\n## 21. Cómo llegan estos datos a Flet\n| Paso | Apartado | Datos / operación |\n|---|---|---|\n| 1 | PopulationEditor | fields contiene TextField; field.value es texto ingresado |\n| 2 | PopulationEditor.save | float convierte cada entrada, tuple empaqueta coordenadas |\n| 3 | Dashboard.settings | lee controles y construye AlgorithmConfig |\n| 4 | Dashboard.apply | await asyncio.to_thread(engine.run,config) ejecuta cálculo fuera del hilo de interfaz |\n| 5 | validate_generations | comprueba estructura y finitud de salida |\n| 6 | Dashboard.snapshots | conserva lista de Generation ya calculada |\n| 7 | Dashboard.render | index elige una generación y scores identifica mejor/promedio |\n| 8 | Chart.set_image | Image.src recibe bytes PNG de gráfica |\n| 9 | page.update | comunica propiedades modificadas al cliente Flet |\n\n### Propiedades de controles relacionadas\n`TextField.value` es texto, no el float del cromosoma. `Dropdown.value` es una clave elegida, como 'torneo'. `Slider.value` es número: mutación visual 5 se convierte a 0.05 mediante división por 100. `Button.on_click` y `TextField.on_submit` guardan callbacks; pasar método sin paréntesis lo registra sin ejecutarlo.\n\n`ListView.controls` contiene las filas visibles del editor; son controles, no la población numérica. `Image.src` contiene imagen codificada, no posiciones. `visible` y `disabled` son estados visuales. `page.update()` transmite modificaciones del árbol de controles; DemoEngine.run no debe llamarlo ni importar Flet.\n\n`async def` crea coroutine; `await` espera sin bloquear el bucle cuando la operación lo permite. `asyncio.to_thread` recibe **referencia de función** y argumentos: no escribas `to_thread(engine.run(config))`, porque ejecutarías el cálculo antes de pasarlo al hilo. La reproducción usa asyncio.sleep para recorrer resultados ya calculados, no ejecuta el motor por cada frame.\n\n## 22. Dónde introducir tu algoritmo real\nConserva AlgorithmConfig y Generation. Reemplaza DemoEngine por una clase con run(config) que aplique selección, cruce y mutación. En cada iteración guarda la población resultante y sus evaluaciones. initial_population puede seguir sirviendo para generar entradas.\n\nEn esta demostración cada posición se calcula **desde la población inicial** y progress; no desde la población anterior. Es interpolación ilustrativa. Un algoritmo genético normalmente produce la siguiente población mediante operadores sobre individuos actuales. No confundir el mínimo muestreado target con una selección genética.\n\n## 23. Guía rápida para leer expresiones compactas\n1. Identifica estructura exterior: lista, tuple o llamada a función.\n2. Busca el for exterior y qué representa cada elemento.\n3. Si hay otro for, identifica si es anidado en otra lista o si aplana el resultado.\n4. Descompón el cuerpo en variables locales: coordenada, individuo, población, puntuación.\n5. Expande la expresión con lista vacía y append; verifica con assert que ambas versiones coinciden.\n6. Distingue referencia de función de ejecución, repetición de secuencia de expansión de argumentos y dato numérico de control Flet.\n\nEstos son recursos de Python, no sintaxis especial de Flet. Los mismos patrones se pueden usar sin una interfaz gráfica.\n")
md('## 3. Parámetros, argumentos y atributos\n`def initial_population(size: int, objective: Objective) -> list[Individual]:`\n\n- `def`: define una función.\n- `size` y `objective`: parámetros, nombres que recibirán valores.\n- `: int` y `: Objective`: anotan los tipos esperados.\n- `-> list[Individual]`: anota el resultado.\n- `:` final y sangría: delimitan el cuerpo.\n- `initial_population(3, objective)`: llamada con argumentos posicionales.\n\n`objective.bounds` accede al atributo bounds del objeto. No es un método porque no tiene `()`. `objective.evaluate(vector)` obtiene la función guardada y la ejecuta con ese vector.\n\n### Objective: todos sus campos\n| Campo | Contenido | Uso |\n|---|---|---|\n| name | texto legible | nombre en interfaz |\n| expression | fórmula como texto | descripción visual |\n| labels | tupla de etiquetas | nombres x,y del editor y ejes |\n| bounds | tupla de pares inferior/superior | generar y validar coordenadas |\n| evaluate | función que recibe Individual | calcular puntuación |\n| dimensions | propiedad calculada con len(labels) | saber cuántas coordenadas hay |\n\n`@dataclass(frozen=True)` en contracts.py genera el constructor y evita reasignar campos. `@property` permite usar `objective.dimensions` en lugar de `objective.dimensions()`. frozen no vuelve inmutable cualquier objeto anidado; en nuestros contratos se usan tuplas para los datos vectoriales.\n')
md('## 4. Empaquetado y desempaquetado de valores\n**Empaquetar** agrupa valores: `point = x, y` produce una tupla aunque no escribas paréntesis. **Desempaquetar** los distribuye: `x, y = point` asigna el primero a x y el segundo a y.\n\nEn `for lower, upper in objective.bounds`, cada vuelta obtiene un par y lo desempaqueta. Equivale a obtener bound y después lower=bound[0], upper=bound[1]. Los dos nombres no representan dos bucles: un único bucle recorre pares.\n\nLos niveles importan: `(xmin,xmax), (ymin,ymax) = bounds` desempaqueta primero dos dominios y después cada par. Sin estrella, la cantidad de valores debe coincidir; si no, se produce ValueError.\n')
md('## 5. Multiplicar listas y tuplas: repetición, no escalado numérico\n`[1,2] * 3` repite la secuencia tres veces. `(1,2) * 3` hace lo mismo conservando tipo tupla. No produce `[3,6]`. Para multiplicar los números, usa `[v*3 for v in values]`.\n\nEn objectives.py, `(domains[key],) * dimensions` repite el par de dominio. La coma crea una tupla **que contiene un par**. Repetirla conserva la estructura `((mín,máx),(mín,máx))`. Es seguro compartir esos pares porque son tuplas de números inmutables.\n\nEn PopulationEditor, `(None,) * dimensions` crea un límite opcional por coordenada. El None de cada posición indica que no hay cotas.\n\n### Peligro con listas anidadas\n`[[0,0]] * 3` repite referencias a la **misma lista interior**. Cambiar una fila cambia las tres. Usa `[[0,0] for _ in range(3)]` para crear tres listas independientes. Una tupla también puede contener una lista mutable; la inmutabilidad exterior no evita modificar esa lista.\n')
md('## 6. Comprensión de listas: un for dentro de corchetes\nSintaxis: `[expresión for variable in iterable]`. Léela como: **para cada variable en iterable, calcula expresión y guarda el resultado en una lista**. El resultado se crea inmediatamente.\n\n`[i*i for i in range(4)]` equivale a crear lista vacía, recorrer range y hacer append(i*i). La expresión aparece primero al escribir, pero al leer conviene empezar por el for.\n\nTambién existe filtro: `[i*i for i in range(4) if i%2 == 0]`. Ese if decide qué elementos se incluyen. No es igual al if/else que elige entre dos valores, explicado después.\n')
md('## 7. Generadores y tuple(...)\n`(expresión for variable in iterable)` crea una **expresión generadora**: entrega valores cuando se le piden. No crea una tupla automáticamente. `tuple(generador)` consume esos valores y crea la tupla.\n\n`tuple(function(v) for v in values)` permite omitir los paréntesis adicionales del generador cuando es el único argumento. Equivale a `tuple((function(v) for v in values))`.\n\nUn generador se agota: recorrerlo una segunda vez no vuelve al principio. Una lista o tupla sí puede recorrerse varias veces. En este motor los generadores son cortos y se materializan inmediatamente con tuple, por lo que las generaciones finales contienen datos reutilizables.\n')
md('## 8. initial_population: validación y aleatoriedad\n`if not 2 <= size <= 200:` es una comparación encadenada: acepta size entre 2 y 200 inclusive. `not` invierte la condición. `raise ValueError(...)` interrumpe la función cuando el tamaño es inválido. La interfaz ya convierte el campo a entero; esta función no comprueba todos los tipos posibles por sí sola.\n\n`rng = random.Random(42)` crea un objeto generador local. 42 es la semilla: repetir esta llamada con igual tamaño y dominio produce la misma muestra dentro del mismo entorno. No es un porcentaje ni una coordenada. Cada llamada a uniform consume un nuevo valor del generador.\n\n`rng.uniform(lower,upper)` devuelve un float entre los límites. El objeto rng tiene estado interno; no reinicies la semilla para cada coordenada porque repetirías valores innecesariamente.\n\n### Descomponer la línea compacta\n```python\nreturn [tuple(rng.uniform(lower, upper) for lower, upper in objective.bounds)\n        for _ in range(size)]\n```\n\n1. El for exterior repite size veces: una vez por individuo.\n2. `_` es un nombre de variable convencional para un valor que no utilizaremos; no es una palabra reservada.\n3. El generador interior recorre cada dominio y desempaqueta lower/upper.\n4. uniform obtiene una coordenada por dominio.\n5. tuple reúne esas coordenadas en un individuo.\n6. Los corchetes exteriores reúnen los individuos en una lista.\n\n`range(size)` produce enteros 0 hasta size−1. El límite superior se excluye. Son índices de repetición, no coordenadas.\n')
md('## 9. Clase DemoEngine, self y run\n`class DemoEngine:` define un tipo que agrupa un método. `engine = DemoEngine()` crea una instancia. No define __init__, por lo que usa la inicialización básica heredada de object. El motor no guarda config como atributo; las variables de run son locales a cada llamada.\n\n`def run(self, config: AlgorithmConfig) -> list[Generation]:` define un método. self es la instancia recibida automáticamente. `engine.run(config)` equivale conceptualmente a `DemoEngine.run(engine,config)`.\n\nLa clase cumple el contrato EvolutionEngine porque proporciona run con la entrada y salida esperadas. No es necesario escribir herencia explícita del Protocol. Protocol describe una interfaz estructural; no implementa selección o mutación por ti.\n\n### Todos los campos de config\n| Atributo | Papel |\n|---|---|\n| population | tupla de vectores iniciales |\n| objective | objeto con evaluate, dominios y etiquetas |\n| mutation_rate | probabilidad [0,1], disponible para tu futuro algoritmo |\n| selection | clave torneo/ruleta/elitista, disponible para tu futuro algoritmo |\n| generations | iteraciones positivas; se agrega generación 0 |\n\nEn DemoEngine, mutation_rate y selection **no se usan para producir cambios**. run presupone configuración válida; la interfaz valida generaciones entre 1 y 500 y población antes de llamarlo. Una llamada manual con generations=0 produciría división entre cero: debes validar antes de usarla.\n\n## 10. axes: comprensión de listas dentro de otra lista\n```python\naxes = [[lower+(upper-lower)*i/60 for i in range(61)]\n        for lower, upper in objective.bounds]\n```\n\nEl for exterior crea una lista por dimensión. El interior crea 61 muestras en el dominio de esa dimensión. El resultado es una lista **de listas**, no una única lista plana.\n\n`upper-lower` mide amplitud; `i/60` es progreso entre 0 y 1. `/` produce división real; `//` sería división entera hacia abajo. `lower + amplitud*progreso` coloca el punto dentro del dominio. Los paréntesis hacen que la resta se calcule antes de multiplicar. Como range(61) incluye 0 y 60, se incluyen ambos extremos y hay 60 intervalos entre 61 puntos.\n')
```

## Flet: constantes y métodos
ThemeMode.LIGHT tema; FontWeight.W_600 grosor medio; ScrollMode.AUTO desplazamiento; CrossAxisAlignment.STRETCH ocupa ancho; BoxFit.CONTAIN mantiene proporción; Icons.* símbolos Material. Border.all(1,color) crea borde uniforme. Control.page sólo está disponible cuando el control fue montado; acceder antes causa RuntimeError. ft.run(main) inicia la app; await page.window.close() cierra prueba. _trigger_event(name,data) es API interna usada sólo por el verificador para despachar clics con cliente real, no por la app.

## Renderización y prueba de escritorio
io.BytesIO() almacena PNG en memoria; getvalue() devuelve bytes. base64.b64decode(text) decodifica PNG mínimo de inicio. matplotlib.figure.Figure configura figura; matplotlib.backends.backend_agg.FigureCanvasAgg la renderiza; matplotlib.colors.LinearSegmentedColormap crea paleta. Las llamadas y argumentos de axes.plot/scatter/grid/labels/imshow, spines, colorbar y print_png se describen en el capítulo 04 y en el inventario exacto de charts.py. zip(*points) separa coordenadas para Matplotlib.

tools/check_desktop.py usa Screenshot para capturar cliente real en 900×700 y 760×640. screenshot(name) espera frame, llama capture con asyncio.wait_for(...,timeout=15), valida cabecera PNG y usa Path.write_bytes para guardar. traceback.format_exc() conserva error; page.on_error recopila errores del cliente. La prueba hace clics por _trigger_event, reproduce, vuelve durante reproducción y alterna vistas repetidamente. Capturas y report.json se guardan en artifacts/desktop_check; failure.txt se escribe si falla. Estas comprobaciones requieren abrir una ventana real.

## Python: operaciones usadas
float/int/str convierten; len cuenta; range produce índices; min/max límites o extremos (key especifica criterio); sum suma; round redondea; enumerate empareja índice/valor; zip empareja listas; list/tuple/set materializan colecciones; any/all evalúan condiciones; sorted ordena; next obtiene primer elemento; isinstance comprueba tipo; getattr obtiene atributo; print muestra resultados; repr conserva representación exacta.

append agrega elemento; extend agrega varios; pop(index) elimina; insert(index,value) inserta; clear vacía; splitlines(keepends=True) conserva saltos; join concatena; strip elimina blancos; startswith comprueba prefijo. ValueError(message) indica dato inválido; TypeError tipo incompatible; RuntimeError estado incompatible; Exception agrupa errores del motor; AssertionError señala prueba fallida. assert comprueba condición. F-strings :g/:.5g cifras significativas, :02d índice, :.0% porcentaje; operadores / de Path unen rutas.

exec(source,namespace) ejecuta celdas de los notebooks propios revisados; no se recomienda aplicarlo a notebooks externos sin revisar. __file__ identifica herramienta. Los campos Jupyter cell_type/metadata/source/outputs/execution_count/id describen celdas; nbformat/nbformat_minor versión 4.5; kernelspec describe kernel.

## Herramientas
md(text) crea celda Markdown; code(text) celda Python; save(index,title,cells) serializa capítulo. check_editor() comprueba CRUD; check_models() contratos; check_dashboard() integración asíncrona; check_notebooks() enlaces y ejercicios. Los métodos privados usan prefijo _ por convención; las definiciones, parámetros y llamadas están incluidos en los bloques de cada módulo.
