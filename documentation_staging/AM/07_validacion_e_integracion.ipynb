## Pruebas de navegación y deprecaciones
Las comprobaciones cubren bloqueo de Resultados sin datos, navegación por botones, Volver, edición pendiente conservada y rutas inválidas. `python -W error::DeprecationWarning tools/check_interface.py` convierte advertencias en errores para detectar usos obsoletos.

## Límites
La demostración no implementa genética; el mínimo es aproximado; puede haber aumento temporal de puntuaciones. La superficie 3D rota mediante sliders y se renderiza a imagen, sin una escena 3D manipulable directamente con el ratón. No hay persistencia, streaming ni exportación. El visor conserva zoom al cambiar imagen. Las gráficas minimizan, y el contrato conserva tamaño de población. Invalidar cálculo no detiene el hilo de un motor externo.
