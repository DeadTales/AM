## Curvas
plot(curve,dots,color=ACCENT,xlabel='x') recibe curva no vacía y puntos (x,y), y devuelve bytes PNG. xlabel nombra el eje horizontal; el historial utiliza 'Generación'. Matplotlib dibuja el gráfico en memoria usando Agg, sin abrir ventanas auxiliares ni usar pyplot.

_figure(height=320) crea Figure(figsize=(8,height/100),dpi=100,layout='constrained',facecolor='white') y axes con add_subplot(111). figsize mide pulgadas; dpi resolución; constrained reserva espacio para etiquetas. tick_params ajusta colores/tamaño; spines son contornos. axes.plot dibuja curva y scatter puntos; grid rejilla; set_xlabel/set_ylabel nombran ejes.

_png(figure) crea BytesIO, renderiza con FigureCanvasAgg(figure).print_png(stream), obtiene stream.getvalue y limpia figure.clear. BytesIO es un archivo binario en memoria. Esto reemplaza cientos de primitivas Canvas por una propiedad src de imagen, reduciendo el trabajo de parches de Flet.

## Mapa de calor
heatmap(objective,population,best) requiere dos variables y devuelve PNG. Muestrea una malla 40×40. LinearSegmentedColormap.from_list(name,colors) crea la paleta clara/azul. imshow(values,extent=(xmin,xmax,ymin,ymax),origin='lower',aspect='auto',cmap=palette) transforma la matriz en mapa; extent fija dominio y origin orientación. scatter añade individuos y el mejor en terracota. figure.colorbar(surface,ax=axes,pad=0.025) explica los colores. set_label y tick_params configuran leyenda. Oscuro significa mayor valor, no mejor en minimización.

La entrada 2D es (x,y), no (x,f(x)): la evaluación se representa por color. Si necesitas una superficie 3D rotatoria, añade otra visualización, conservando el contrato del motor. No es necesaria para editar ni recorrer individuos 2D.

## Superficie tridimensional
Las funciones con dos entradas admiten superficie 3D como opción predeterminada. surface3d(objective,population,best,elevation=30,azimuth=-55) dibuja z=f(x,y) y devuelve PNG. elevation y azimuth son grados de cámara; el mejor individuo aparece en terracota. NumPy.linspace(start,stop,num) muestrea 55 posiciones por eje, meshgrid crea matrices x/y y array convierte evaluaciones a matriz z.

Figure.add_subplot(111,projection='3d',computed_zorder=False) crea ejes tridimensionales. plot_surface(x,y,z,cmap,alpha,linewidth,antialiased,rcount,ccount,zorder) define superficie: alpha transparencia, rcount/ccount resolución y zorder orden de dibujo. scatter(x,y,z,depthshade=False,...) añade individuos; sus alturas son evaluaciones, no genes adicionales. view_init(elev,azim) orienta cámara y set_box_aspect((1,1,0.65)) controla proporciones. Los marcadores se dibujan por delante para distinguir la población, incluso cuando el relieve la taparía.

Dashboard.chart_mode elige surface/heatmap; chart_controls se muestra sólo en dos variables y camera_controls sólo en superficie. azimuth y elevation son sliders con on_change_end: se redibuja al soltarlos para evitar cálculos continuos. change_chart(e=None) conserva generación y llama _render_function(generation,best), sin recalcular el motor. El zoom/arrastre del visor transforma la imagen; la cámara se cambia con los sliders, no arrastrando la superficie.

## Chart persistente
Chart(height=320) hereda Container. Su ancho ocupa el disponible y su alto es explícito. raster es Image(src=bytes,width=800,height=height,fit=CONTAIN,gapless_playback=True): src contiene PNG; width/height proporcionan tamaño inicial, fit mantiene proporción y gapless_playback conserva imagen anterior mientras se carga la nueva. Una imagen mínima inicial evita contenido invisible, porque InteractiveViewer exige content visible.

viewer es InteractiveViewer(content=raster,constrained=True,min_scale=1,max_scale=4,trackpad_scroll_causes_scale=True,height=height,width=infinito). constrained aplica las restricciones disponibles al hijo. set_image(png) cambia src conservando visor y zoom. No utiliza callbacks on_resize que reconstruyan dibujos. Los valores exactos están en la lista de resultados; no hay tooltips de puntos.

La gráfica de historial muestra todas las generaciones recibidas y marca la visible. No significa ejecución en streaming ni garantiza mejora monótona.
